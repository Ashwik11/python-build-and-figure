# Create a Mini E-Commerce Dataset in Python

A list of dictionaries is a common way to represent a small dataset in Python. The list stores multiple records, while each dictionary describes one record with meaningful keys.

In this tutorial, we will create and update a small product catalog without using loops.

## What You Will Learn

By the end of this tutorial, you will know how to:

- Combine a list with dictionaries
- Store multiple product records
- Select records using list indexes
- Select fields using dictionary keys
- Update an existing field
- Add a new field to one product
- Add another product manually
- Understand the structure of a small dataset

## 1. Create the Product Dataset

The outer list contains three dictionaries. Each dictionary represents one product and follows the same structure: `name`, `price`, and `brand`.

```python
products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell"
    },
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung"
    },
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple"
    }
]
```

## Understanding the Structure

The dataset has two levels:

1. The **list** stores all product records.
2. Each **dictionary** stores the fields for one product.

The general access pattern is:

`products[list_index]["dictionary_key"]`

The list index chooses a product. The dictionary key chooses a field from that product.

## Dataset Map

| List index | Product | Available keys |
|---:|---|---|
| `0` | Laptop | `name`, `price`, `brand` |
| `1` | Phone | `name`, `price`, `brand` |
| `2` | Tablet | `name`, `price`, `brand` |

Python indexing begins at `0`, so the first product is stored at index `0`.

## 2. Print All Products

Printing the outer list displays the complete dataset. We can also print each original product directly to make the three records easy to read without using a loop.

```python
print("Complete product dataset:", products)
print("Product 1:", products[0])
print("Product 2:", products[1])
print("Product 3:", products[2])
```

## 3. Print the First Product

Using one list index returns the complete dictionary stored at that position. Index `0` returns the first product.

```python
print("First product:", products[0])
```

## 4. Print the Second Product's Price

The index `1` selects the second product dictionary. The key `price` then selects its price.

```python
print("Second product's price:", products[1]["price"])
```

## 5. Print the Third Product's Brand

The index `2` selects the third product dictionary. The key `brand` then retrieves its brand.

```python
print("Third product's brand:", products[2]["brand"])
```

## 6. Change the First Product's Price

Lists and dictionaries are mutable. We can select the first product and assign a new value to its `price` key.

We will change the laptop price from `70000` to `75000`.

```python
products[0]["price"] = 75000
print("Updated first product:", products[0])
```

## 7. Add a Rating to the Second Product

Assigning a value to a new dictionary key adds a field. We will add a `rating` of `4.5` to the phone record.

This also demonstrates that dictionaries in the same list do not have to contain exactly the same keys, although consistent schemas are usually easier to process.

```python
products[1]["rating"] = 4.5
print("Updated second product:", products[1])
```

## 8. Add Another Product Manually

The `append()` method adds one new element to the end of a list. Our new element is another complete product dictionary.

We will add Sony headphones as the fourth product.

```python
products.append({
    "name": "Headphones",
    "price": 15000,
    "brand": "Sony"
})

print("Newly added product:", products[3])
```

## 9. Print the Final Dataset

The final dataset contains four products, the updated laptop price, and the phone rating.

```python
print("Final product dataset:", products)
print("Final product 1:", products[0])
print("Final product 2:", products[1])
print("Final product 3:", products[2])
print("Final product 4:", products[3])
```

## List, Dictionary, Indexing, Keys, and Values

### List

The outer list stores multiple product records in a defined order. Each product has a numerical index.

### Dictionary

Each dictionary represents one product and connects descriptive keys to values.

### Indexing

A list index selects a product: `products[1]`.

### Keys

Dictionary keys such as `name`, `price`, and `brand` describe each field.

### Values

Values are the actual product information, such as `Phone`, `40000`, and `Samsung`.

Combining both forms gives `products[1]["price"]`: select the second product, then retrieve its price.

## Access Pattern Summary

| Expression pattern | Result |
|---|---|
| `products` | Complete dataset |
| `products[index]` | Complete product dictionary |
| `products[index]["name"]` | Product name |
| `products[index]["price"]` | Product price |
| `products[index]["brand"]` | Product brand |
| `products[index]["key"] = value` | Adds or updates a field |
| `products.append(dictionary)` | Adds another product record |

## Complete No-Loop Verification

The following independent cell recreates the original dataset and performs every required operation without using loops.

```python
products = [
    {"name": "Laptop", "price": 70000, "brand": "Dell"},
    {"name": "Phone", "price": 40000, "brand": "Samsung"},
    {"name": "Tablet", "price": 30000, "brand": "Apple"}
]

print("All products:", products)
print("First product:", products[0])
print("Second product's price:", products[1]["price"])
print("Third product's brand:", products[2]["brand"])

products[0]["price"] = 75000
products[1]["rating"] = 4.5
products.append({"name": "Headphones", "price": 15000, "brand": "Sony"})

print("Updated first product price:", products[0]["price"])
print("Second product rating:", products[1]["rating"])
print("Added product:", products[3])
print("Final dataset:", products)
```
