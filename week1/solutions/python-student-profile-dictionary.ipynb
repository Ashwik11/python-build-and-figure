# Build a Student Profile with a Python Dictionary

Python dictionaries store related information as key-value pairs. Descriptive keys make dictionaries useful for profiles, product records, configuration settings, and API data.

In this tutorial, we will create, inspect, update, remove, retrieve, and copy student profile data.

## What You Will Learn

By the end of this tutorial, you will know how to:

- Create and print a dictionary
- Access values using keys
- Inspect keys, values, and key-value pairs
- Update existing information
- Add new fields
- Remove a field with `pop()`
- Retrieve information safely with `get()`
- Create a separate dictionary with `copy()`

## 1. Create and Print the Student Dictionary

The dictionary contains five fields: name, age, course, city, and marks. Each key describes the value associated with it.

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}

print("Complete student dictionary:", student)
```

## Student Profile Structure

| Key | Value | Data type |
|---|---|---|
| `name` | Rahul | String |
| `age` | 22 | Integer |
| `course` | Python | String |
| `city` | Bangalore | String |
| `marks` | 88 | Integer |

## 2. Print the Student's Name and Course

A dictionary value can be accessed by placing its key inside square brackets.

```python
print("Student name:", student["name"])
print("Course:", student["course"])
```

## 3. Print All Keys — `keys()`

The `keys()` method returns a dynamic view containing every dictionary key. Converting it to a list produces a familiar list representation.

```python
print("All keys:", student.keys())
print("Keys as a list:", list(student.keys()))
```

## 4. Print All Values — `values()`

The `values()` method returns a dynamic view containing every value in the dictionary.

```python
print("All values:", student.values())
print("Values as a list:", list(student.values()))
```

## 5. Print All Key-Value Pairs — `items()`

The `items()` method returns every entry as a `(key, value)` tuple.

```python
print("All key-value pairs:", student.items())
print("Items as a list:", list(student.items()))
```

## 6. Change the Marks with `update()`

The `update()` method accepts another dictionary or key-value data. Existing keys are updated, while new keys are added.

We will change Rahul's marks from `88` to `92`.

```python
student.update({"marks": 92})
print("Updated marks:", student["marks"])
```

## 7. Add Email and Phone Information

A single `update()` call can add multiple new key-value pairs. Phone numbers are commonly stored as strings because they may contain country codes or formatting symbols and are not used for arithmetic.

```python
student.update({
    "email": "rahul@example.com",
    "phone": "+91-9876543210"
})

print("Email:", student["email"])
print("Phone:", student["phone"])
```

## 8. Remove the City — `pop()`

The `pop()` method removes a specified key and returns its associated value. We will store the removed city so that we can display it.

Calling `pop()` without a default value raises a `KeyError` if the key does not exist.

```python
removed_city = student.pop("city")

print("Removed city:", removed_city)
print("Profile after removing city:", student)
```

## 9. Retrieve the Name Safely — `get()`

The `get()` method retrieves the value associated with a key. Unlike square-bracket access, it does not raise a `KeyError` when the key is missing.

An optional second argument can provide a default value for a missing key.

```python
student_name = student.get("name")
missing_grade = student.get("grade", "Not provided")

print("Name retrieved with get():", student_name)
print("Missing grade with a default value:", missing_grade)
```

## 10. Create a Dictionary Copy — `copy()`

The `copy()` method creates a new dictionary containing the same top-level entries. Changing a top-level value in the original will not change the copied dictionary.

This is a shallow copy, so nested mutable objects would still be shared unless a deep copy were used. Our current student dictionary contains only simple top-level values.

```python
student_copy = student.copy()

print("Original student dictionary:", student)
print("Copied student dictionary:", student_copy)
print("Are they different dictionary objects?", student is not student_copy)
```

## Dictionary Method Summary

| Method | Purpose | Important behavior |
|---|---|---|
| `keys()` | Returns all keys | Produces a dynamic dictionary view |
| `values()` | Returns all values | Produces a dynamic dictionary view |
| `items()` | Returns key-value pairs | Each entry appears as a tuple |
| `get()` | Retrieves a value safely | Can return a default when a key is missing |
| `update()` | Adds or updates entries | Can process multiple entries at once |
| `pop()` | Removes a key | Returns the removed value |
| `copy()` | Creates a shallow copy | Produces a separate outer dictionary |

## Square Brackets vs. `get()`

Both approaches retrieve dictionary values:

- `student["name"]` returns the value but raises `KeyError` when the key is absent.
- `student.get("name")` returns the value and returns `None` when the key is absent.
- `student.get("grade", "Not provided")` returns a custom default when the key is absent.

## Complete Verification

The following independent cell re-creates the original profile and performs every required operation in order.

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}

print("Complete dictionary:", student)
print("Student name:", student["name"])
print("Course:", student["course"])
print("All keys:", student.keys())
print("All values:", student.values())
print("All key-value pairs:", student.items())

student.update({"marks": 92})
student.update({
    "email": "rahul@example.com",
    "phone": "+91-9876543210"
})
removed_city = student.pop("city")
retrieved_name = student.get("name")
student_copy = student.copy()

print("Updated marks:", student["marks"])
print("Email:", student["email"])
print("Phone:", student["phone"])
print("Removed city:", removed_city)
print("Name retrieved with get():", retrieved_name)
print("Updated student dictionary:", student)
print("Copied dictionary:", student_copy)
```
