# Python Set Operations: Compare Student Groups

Sets are useful when we need to compare groups, find shared members, identify differences, or keep only unique values.

In this tutorial, we will compare students learning Python and Java using Python's fundamental set operations.

## What You Will Learn

By the end of this tutorial, you will know how to:

- Create and print Python sets
- Combine groups with `union()`
- Find shared members with `intersection()`
- Find members exclusive to one group with `difference()`
- Find members belonging to exactly one group with `symmetric_difference()`
- Add and remove set elements
- Understand the difference between `remove()` and `discard()`

## 1. Create the Student Sets

We will create two sets: one for Python students and one for Java students.

A set stores unique values and does not guarantee display order. Therefore, the names may appear in a different order when the notebook is run.

```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

print("Python students:", python_students)
print("Java students:", java_students)
```

## 2. Students Learning Either Python or Java — `union()`

The `union()` method combines both sets and returns every unique student who belongs to at least one group. Students appearing in both sets are included only once.

```python
either_language = python_students.union(java_students)
print(either_language)
```

## 3. Students Learning Both — `intersection()`

The `intersection()` method returns values that are present in both sets. These students are learning both Python and Java.

```python
both_languages = python_students.intersection(java_students)
print(both_languages)
```

## 4. Students Learning Only Python — `difference()`

`python_students.difference(java_students)` returns students found in the Python set but not in the Java set.

The order of the two sets matters when using `difference()`.

```python
print(python_students)
print(java_students)
only_python = python_students.difference(java_students)
print(only_python)
```

## 5. Students Learning Only Java — `difference()`

Reversing the sets returns students found in the Java set but not in the Python set.

```python
print(python_students)
print(java_students)
only_java = java_students.difference(python_students)
print(only_java)
```

## 6. Students in Exactly One Group — `symmetric_difference()`

The `symmetric_difference()` method returns students who belong to exactly one set. It excludes students who are present in both groups.

This is equivalent to combining the only-Python and only-Java results.

```python
exactly_one_group = python_students.symmetric_difference(java_students)
print(exactly_one_group)
```

## 7. Add a New Student — `add()`

The `add()` method inserts one element into a set. We will add `Vikram` to the Python group.

If the value already exists, the set remains unchanged because sets cannot contain duplicates.

```python
print(python_students)
python_students.add("Vikram")
print(python_students)
```

## 8. Remove a Student — `remove()`

The `remove()` method deletes a specified element. We will remove `Rohit` from the Java group.

If the specified value does not exist, `remove()` raises a `KeyError`.

```python
print(java_students)
java_students.add("Rohit")
java_students.remove("Rohit")
print(java_students)
```

## 9. Safely Remove a Value — `discard()`

The `discard()` method removes a specified element if it exists. Unlike `remove()`, it does not raise an error when the value is absent.

We will attempt to discard `Anjali`, who is not currently in the Python set. The operation completes safely and leaves the set unchanged.

```python
java_students.discard("Rohit")
print(java_students)
```

## Set Method Summary

| Method | Purpose | Key behavior |
|---|---|---|
| `union()` | Combines sets | Keeps every unique value |
| `intersection()` | Finds common values | Keeps values present in both sets |
| `difference()` | Finds values unique to the first set | The order of the operands matters |
| `symmetric_difference()` | Finds values in exactly one set | Excludes common values |
| `add()` | Adds one value | Does nothing if the value already exists |
| `remove()` | Removes a value | Raises `KeyError` if absent |
| `discard()` | Safely removes a value | Does nothing if absent |

## `remove()` vs. `discard()`

Both methods remove a value from a set, but they differ when the value is missing:

- Use `remove()` when a missing value should be treated as an error.
- Use `discard()` when the value may be absent and the program should continue safely.
