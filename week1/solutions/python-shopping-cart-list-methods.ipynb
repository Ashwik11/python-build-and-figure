# Build a Shopping Cart with Python Lists

In this tutorial, we will build a simple shopping cart and use it to explore the most important Python list operations.

## 1. Create the Shopping Cart

We will begin by creating a variable named `cart`.

The list will contain five product names:

- Laptop
- Mouse
- Keyboard
- Monitor
- Headphones

A Python list is written using square brackets, with commas separating its elements.

```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

## 2. Display All Products

The `print()` function can display the complete shopping cart.

Because a list preserves the order of its elements, the products will appear in the same order in which they were added.

```python
print(cart)
```

## Understanding List Indexing

Every element in a Python list has an index.

Python uses zero-based indexing:

| Index | Product |
|---:|---|
| `0` | Laptop |
| `1` | Mouse |
| `2` | Keyboard |
| `3` | Monitor |
| `4` | Headphones |

Negative indexes count backward from the end:

- `-1` represents the last item.
- `-2` represents the second-to-last item.

## 3. Access the First and Last Products

We will use positive indexing to access the first product and negative indexing to access the last product.

- Index `0` returns the first product.
- Index `-1` returns the last product.

```python
print(cart[0])
print(cart[-1])
```

## 4. Add `Webcam` with `append()`

The `append()` method adds one element to the end of a list.

We will use it to add `Webcam` to the shopping cart.

```python
cart.append("Webcam")
print(cart)
```

## 5. Insert `USB Hub` at Index 2

The `insert()` method adds an element at a specific index.

Its general syntax is:

`list.insert(index, value)`

We will insert `USB Hub` at index `2`. The existing elements at and after that index will move one position to the right.

```python
cart.insert(2, "USB Hub")
```

```python
print(cart)
```

## 6. Remove `Mouse` by Value

The `remove()` method deletes the first occurrence of a specified value.

We will remove `Mouse` from the cart by providing its value rather than its index.

```python
cart.remove("Mouse")
print(cart)
```

## 7. Remove the Last Product with `pop()`

The `pop()` method removes and returns an element.

When no index is provided, it removes the last item in the list.

Because `Webcam` was appended to the end of the cart, and no later operation moved it, it should be the item removed by `pop()`.

```python
removed_element = cart.pop()
#print(removed_element)
print(cart)
```

## 8. Find the Index of `Monitor`

The `index()` method searches for a value and returns the index of its first occurrence.

The cart has already been modified, so the current index of `Monitor` may differ from its position in the original list.

```python
cart.index("Monitor")
```

## 9. Count the Occurrences of `Laptop`

The `count()` method returns the number of times a specified value appears in a list.

We will use it to determine how many times `Laptop` occurs in the shopping cart.

```python
cart.count("Laptop")
cart_copy = cart.copy()
cart_copy.append("Laptop")
cart_copy.count("Laptop")
```

## 10. Create a Copy of the Cart

The `copy()` method creates a new list containing the same elements as the original list.

The copied list has its own list identity. Changes to the outer structure of the original list will not automatically change the copied list.

We will create the copy before reversing and sorting the original cart so that we can compare them.

```python
cart_copy = cart.copy()
```

## Assignment Is Not the Same as Copying

Writing:

`second_cart = cart`

does not create an independent list. Both variables refer to the same list.

Using:

`second_cart = cart.copy()`

creates a separate list containing the same elements.

For this tutorial, we will use `copy()` so that the saved cart remains unchanged when the original cart is reordered.

```python
cart_copy = cart.copy()
```

## 11. Reverse the Cart

The `reverse()` method reverses the current order of the list in place.

It does not arrange the products alphabetically. It simply changes the first item to the last, the second item to the second-to-last, and so on.

```python
cart.reverse()
print(cart)
cart.reverse()
print(cart)
```

## 12. Sort the Products Alphabetically

The `sort()` method arranges the elements of a list in ascending order.

Because our list contains strings, ascending order means alphabetical order.

The method changes the original list in place.

```python
print(cart)
cart.sort()
print(cart)
```

## Comparing Python List Methods

| Method | Purpose | Important Behavior |
|---|---|---|
| `append(value)` | Adds one element to the end | The provided value becomes one list element |
| `extend(iterable)` | Adds multiple elements to the end | Each value from the iterable is added separately |
| `insert(index, value)` | Adds an element at a selected index | Existing elements move to the right |
| `remove(value)` | Removes an element by value | Removes the first matching value |
| `pop()` | Removes an element by index | Removes the last element when no index is supplied and returns it |
| `clear()` | Removes every element | Leaves an empty list |
| `copy()` | Creates a separate list copy | The new outer list can be modified independently |
| `sort()` | Arranges elements in ascending order | Changes the list in place |
| `reverse()` | Reverses the current order | Does not sort the values |

## `append()` vs. `extend()`

Both methods add elements to the end of a list, but they behave differently.

### `append()`

`append()` adds its argument as one element.

If a list is appended to another list, the entire added list becomes one nested element.

### `extend()`

`extend()` takes an iterable and adds each of its elements separately.

Use `append()` when adding one item. Use `extend()` when adding multiple items.

```python
append_example = ["Laptop"]
append_example.append(["Mouse","Keyboard"])

extend_example = ["Laptop"]
extend_example.extend(["Mouse","Keyboard"])

print("using append", append_example)
print("using extend", extend_example)
```

## `remove()` vs. `pop()`

### `remove()`

- Removes an item by its value
- Removes the first matching value
- Raises an error if the value does not exist

### `pop()`

- Removes an item by its index
- Removes the last item when no index is provided
- Returns the removed value
- Raises an error if the index is invalid

## `sort()` vs. `reverse()`

These methods both change the order of a list, but they have different purposes.

### `sort()`

Arranges the values according to their natural ascending order, such as alphabetical or numerical order.

### `reverse()`

Reverses the list's current order without comparing or sorting its values.

## Demonstrating `clear()` Safely

The `clear()` method removes every element from a list.

We will demonstrate it on a temporary copy rather than clearing the main shopping cart. This allows us to preserve the completed result while still understanding how `clear()` works.

```python
extend_example.clear()
print(extend_example)
```

## Common Beginner Mistakes

### Expecting list methods to create a new list

Methods such as `append()`, `insert()`, `remove()`, `sort()`, and `reverse()` modify the existing list.

### Assigning the result of `sort()` or `reverse()`

These methods change the list in place and return `None`. Call the method first, then print the list separately.

### Confusing an index with a value

`remove("Mouse")` removes a value, while `pop(1)` removes the item at index `1`.

### Forgetting that indexes change

Adding or removing products can change the positions of later products.

### Using `clear()` on the main cart too early

Once the cart is cleared, later operations cannot use its products unless the list is recreated.

### Confusing `append()` and `extend()`

`append()` adds one item. `extend()` adds each element from another iterable.
