# Python Nested Dictionaries: Employee Profile Tutorial

A nested dictionary is a dictionary containing another dictionary as one of its values. This structure is useful for organizing related information into clear categories.

In this tutorial, we will create, access, and update an employee profile without using loops.

## What You Will Learn

By the end of this tutorial, you will know how to:

- Create a nested dictionary
- Access top-level dictionary values
- Access values inside a nested dictionary
- Update existing values
- Add new top-level key-value pairs
- Add new values inside a nested dictionary
- Understand chained dictionary indexing

## 1. Create the Employee Dictionary

The employee dictionary contains four top-level keys: `name`, `department`, `skills`, and `salary`.

The value associated with `skills` is another dictionary, making this a nested dictionary.

```python
employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

print("Complete employee dictionary:", employee)
```

## Dictionary Structure

The structure can be understood as two levels:

| Level | Key | Value |
|---|---|---|
| Top level | `name` | Amit |
| Top level | `department` | Engineering |
| Top level | `skills` | Nested skills dictionary |
| Top level | `salary` | 80000 |
| Inside `skills` | `language` | Python |
| Inside `skills` | `database` | PostgreSQL |
| Inside `skills` | `cloud` | AWS |

## 2. Print the Employee Name and Department

Top-level values require one key inside square brackets. For example, `employee["name"]` retrieves the value associated with the `name` key.

```python
print(employee["name"])
print(employee["department"])
```

## 3. Print the Complete Skills Dictionary

The `skills` key points to the complete nested dictionary. Using one key returns that entire inner dictionary.

```python
print(employee["skills"])
```

## 4. Access Individual Nested Skills

To retrieve one skill, we use two keys:

`employee["skills"]["skill_key"]`

The first key retrieves the nested skills dictionary. The second key retrieves a specific value from that dictionary.

```python
print(employee)
print(employee["skills"]["language"])
```

## 5. Change the Programming Language

Dictionaries are mutable, so an existing value can be replaced by assigning a new value to its key.

We will change the nested `language` value from `Python` to `Python + JavaScript`.

```python
employee["skills"]["language"] = "Python + Javascript"
print(employee)
```

## 6. Change the Salary

Salary is stored at the top level, so it requires only one key. We will update it from `80000` to `90000`.

```python
employee["salary"] = 90000
print(employee)
```

## 7. Add Employee Experience

Assigning a value to a key that does not yet exist creates a new key-value pair.

We will add the top-level key `experience` with the integer value `3`.

```python
employee["experience"] = 3
print(employee)
```

## 8. Add Another Nested Skill

A new value can also be added directly to the nested skills dictionary.

We will add the key `framework` with the value `FastAPI`.

```python
employee["skills"]["framework"] = "FASTAPI"
```

## 9. Print the Updated Employee Dictionary

The final dictionary should contain the updated language and salary, the new experience field, and the additional framework skill.

```python
print(employee)
```

## How Nested Dictionaries Are Accessed

Nested dictionaries are accessed one level at a time:

- `employee["name"]` accesses a top-level value.
- `employee["skills"]` accesses the complete nested dictionary.
- `employee["skills"]["language"]` first accesses `skills`, then accesses `language` inside it.

Each pair of square brackets moves one level deeper into the structure.

Using a key that does not exist with square-bracket access raises a `KeyError`. In applications where a key may be absent, the `get()` method can provide safer access.

## Access and Update Summary

| Operation | Pattern |
|---|---|
| Access a top-level value | `dictionary["key"]` |
| Access a nested dictionary | `dictionary["nested_key"]` |
| Access a nested value | `dictionary["nested_key"]["inner_key"]` |
| Update a top-level value | `dictionary["key"] = new_value` |
| Update a nested value | `dictionary["nested_key"]["inner_key"] = new_value` |
| Add a top-level key | Assign a value to a new top-level key |
| Add a nested key | Assign a value to a new key in the nested dictionary |

## Complete No-Loop Verification

The following independent cell re-creates the original employee dictionary and performs every required operation in order without using loops.

```python
employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

print("Employee name:", employee["name"])
print("Department:", employee["department"])
print("Complete skills dictionary:", employee["skills"])
print("Programming language:", employee["skills"]["language"])
print("Database:", employee["skills"]["database"])
print("Cloud technology:", employee["skills"]["cloud"])

employee["skills"]["language"] = "Python + JavaScript"
employee["salary"] = 90000
employee["experience"] = 3
employee["skills"]["framework"] = "FastAPI"

print("Updated language:", employee["skills"]["language"])
print("Updated salary:", employee["salary"])
print("Experience:", employee["experience"])
print("Added framework:", employee["skills"]["framework"])
print("Final employee dictionary:", employee)
```
