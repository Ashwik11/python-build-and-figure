## Why Do Data Types Matter?

Data types tell Python what kind of value a variable contains and what operations can be performed with it.

For example:

- Strings can be joined and transformed.
- Integers and floats can be used in calculations.
- Boolean values can be used when making decisions.

Choosing an appropriate data type makes programs easier to understand and helps prevent unexpected errors.

## Python Data Types Used in This Tutorial

Different kinds of information require different data types. Our developer profile will use four fundamental Python data types:

- **String (`str`):** Stores text, such as a name or city
- **Integer (`int`):** Stores whole numbers, such as age
- **Float (`float`):** Stores decimal numbers, such as years of experience
- **Boolean (`bool`):** Stores either `True` or `False`

We will create at least one variable using each of these data types.

### 1. Name

I will create a variable to store my name.

Because a name contains text, its data type will be a **string (`str`)**.

```python
name = "Ashwik"
```

### 2. Age

I will create a variable to store my age.

Age is normally represented as a whole number, so its data type will be an **integer (`int`)**.

```python
age = 27
```

### 3. City

I will create a variable to store the city where I currently live.

A city name is text, so its data type will be a **string (`str`)**.

```python
city = "Dallas"
```

### 4. College or Company Name

I will create a variable to store the name of my college or company.

The name contains text, so its data type will be a **string (`str`)**.

```python
college = "Belhaven University"
```

### 5. Current Learning Topic

I will create a variable to represent the subject I am currently learning.

The learning topic is written as text, so its data type will be a **string (`str`)**.

```python
current_topic = "Python Programming"
```

### 6. Years of Experience or Study

I will create a variable to store my years of professional experience or study.

I will represent this value using a **float (`float`)** so that it can also support partial years, such as 2.5 years.

```python
years_of_experience = 2.0
```

### 7. Do I Like Python?

I will create a variable to indicate whether I like Python.

This variable will contain either `True` or `False`, so its data type will be a **Boolean (`bool`)**.

```python
likes_python = True
```

### 8. Expected Salary After Learning Python

I will create a variable to store my expected salary after improving my Python skills.

Because a salary can contain decimal values, I will represent it using a **float (`float`)**.

```python
expected_salary = 140000.0
```

## Printing My Developer Profile

Now that all the variables have been created, I will use the `print()` function to display every value.

I will include a descriptive label with each value so that the output is clear and easy to understand.

```python
print(name)
print(age)
print(city)
print(college)
print(current_topic)
print(years_of_experience)
print(likes_python)
print(expected_salary)
```

## Checking the Data Type of Every Variable

Python provides the built-in `type()` function to identify the data type of a value or variable.

In this section, I will use `type()` to display the data type of every variable in my developer profile.

```python
print(type(name))
print(type(age))
print(type(city))
print(type(college))
print(type(current_topic))
print(type(years_of_experience))
print(type(likes_python))
print(type(expected_salary))
```

## Common Beginner Mistakes

### Adding quotation marks around numbers

A value such as `"27"` is a string, while `27` is an integer.

### Using lowercase Boolean values

Python uses `True` and `False` with capital first letters. Writing `true` or `false` will produce an error.

### Using unclear variable names

Names such as `age` and `current_topic` are easier to understand than names such as `a` and `x`.

### Starting a variable name with a number

Python variable names cannot begin with a number. For example, `2years` is invalid, but `years_of_experience` is valid.
