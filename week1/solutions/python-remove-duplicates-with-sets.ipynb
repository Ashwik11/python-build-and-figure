# Remove Duplicate Data Using Python Sets

Real-world data often contains repeated values. Python sets provide a simple way to keep only unique elements.

In this tutorial, we will remove duplicate numbers and student names without using loops. We will also compare the number of original and unique elements.

- Create and print a Python list

- Remove duplicate values without loops
- Convert a set back into a list
- Count original and unique elements with `len()`
- Apply the same technique to duplicate names
- Understand the benefits and limitations of sets

## Lists and Sets

A **list** is an ordered collection that allows duplicate values. A **set** stores only unique values and does not guarantee positional order.

| Feature | List | Set |
|---|---|---|
| Allows duplicates | Yes | No |
| Maintains positional order | Yes | No guaranteed order |
| Supports indexing | Yes | No |
| Useful for removing duplicates | Requires extra work | Yes |

## 1. Create and Print the Original Number List

The list below intentionally contains repeated occurrences of `10`, `20`, and `30`.

```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

print("Original list:", numbers)
```

## 2. Convert the List into a Set

Passing the list to `set()` creates a collection containing one copy of every distinct number.

The duplicate occurrences of `10`, `20`, and `30` disappear. Because sets do not guarantee positional order, the displayed order may differ between environments.

```python
unique_numbers = set(numbers)
print(unique_numbers)
```

## 3. Convert the Set Back into a List

The `list()` function converts the set of unique numbers back into a list. The new list contains no duplicate values.

Converting back to a list restores list operations such as indexing, but it does not restore the original ordering automatically.

```python
unique_numbers_list = list(unique_numbers)
print(unique_numbers_list)
```

## 4. Compare Original and Unique Element Counts

The `len()` function returns the number of elements in a collection.

The original list counts every occurrence, while the set counts only distinct values. The difference tells us how many repeated occurrences were removed.

```python
original_count = len(numbers)
unique_count = len(unique_numbers_list)
removed_count = original_count - unique_count
print(removed_count)
```

## 5. Remove Duplicate Student Names

The same technique works with strings. This example contains repeated student names. Converting the list to a set keeps one occurrence of each name.

```python
student_names = ["Rahul", "Priya", "Aman", "Rahul", "Neha", "Priya", "Aman"]
print(student_names)
set_elements = set(student_names)
print(set_elements)
list_elements = list(set_elements)
print(list_elements)

```

## Why Sets Are Useful for Duplicate Data

Sets are useful because they automatically enforce uniqueness. This makes them helpful for:

- Removing duplicate IDs, names, email addresses, or categories
- Checking whether a value already exists
- Comparing groups of values
- Finding unions, intersections, and differences between datasets

### Important limitation

A set should not be used when the original order or duplicate frequency must be preserved. Sets are also not accessed by numerical index.
