# Python Nested Lists and Indexing

A nested list is a list that contains other lists as its elements. Nested lists are useful for organizing related information into rows and columns.

In this tutorial, we will create a collection of student records and learn how to access, update, and add information using nested-list indexing.

All operations will be completed without using loops.

## What You Will Learn

By the end of this tutorial, you will know how to:

- Create a nested list
- Print a complete nested list
- Access an individual record
- Access a specific value inside a record
- Use positive indexing with nested lists
- Update a value inside a nested list
- Add a new record manually
- Understand row and column positions

## What Is a Nested List?

A nested list is a list containing one or more lists.

In our example:

- The outer list stores all student records.
- Each inner list represents one student.
- Every student record contains a name, age, and course.

This structure is similar to a table containing rows and columns.

## 1. Create the Student Records

We will create a variable named `students`.

It will contain three inner lists:

1. Rahul's record
2. Priya's record
3. Aman's record

Each student record follows this structure:

`[name, age, course]`

```python
students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"]
]
```

## 2. Print the Complete Nested List

We will first print the complete `students` list.

The output will contain all three student records, including their names, ages, and courses.

```python
print(students)
```

## How Nested-List Indexing Works

To access a value in a nested list, we use two indexes:

`students[row_index][column_index]`

- The first index selects a student record from the outer list.
- The second index selects a value from that student's inner list.

Python indexing begins at `0`.

## 3. Access Rahul's Name

Rahul is stored in the first student record, and the name is the first value inside that record.

We will use one index to select Rahul's record and another index to select his name.

```python
students[0][0]
```

## 4. Access Priya's Age

Priya is stored in the second student record.

After selecting Priya's record, we will use the inner-list index for age.


```python
students[1][1]
```

## 5. Access Aman's Course

Aman is stored in the third student record.

After selecting Aman's record, we will use the inner-list index for the course.

```python
students[2][2]
```

## 6. Access Priya's Complete Record

When we provide only the outer-list index, Python returns the complete inner list at that position.

We will select Priya's record without providing a second index.

```python
students[1]
```

## 7. Change Rahul's Course

Python lists are mutable, which means their values can be changed after the list has been created.

We will locate Rahul's course using nested indexing and change it from `Python` to `AI`.

```python
print(students)
students[0][2] = "AI"
print(students)
```

## Understanding the Update

Updating a nested-list value follows the same indexing pattern used to access it:

`students[row_index][column_index]`

The selected location can be assigned a new value because both the outer list and the inner student records are mutable lists.

Only Rahul's course changes. His name and age remain unchanged.

## 8. Add Another Student Record

A new student must follow the same structure as the existing records:

`[name, age, course]`

We will manually create one additional student record and add it to the outer `students` list.

The `append()` method adds one new element to the end of a list.

```python
new_student = ["Neha", 23, "Cloud Computing"]
students.append(new_student)
```

## 9. Print the Updated Nested List

Finally, we will print the complete nested list again.

The updated output should show:

- Rahul's course changed to `AI`
- Priya's original record
- Aman's original record
- The newly added fourth student

```python
print(students)
```

## Nested-List Indexing Summary

| Expression Pattern | Result |
|---|---|
| `students` | The complete nested list |
| `students[row]` | One complete student record |
| `students[row][0]` | The selected student's name |
| `students[row][1]` | The selected student's age |
| `students[row][2]` | The selected student's course |
| `students[row][column] = value` | Updates a specific value |
| `students.append(record)` | Adds another student record |

## Think of the Nested List as a Table

| Row Index | Name — Column 0 | Age — Column 1 | Course — Column 2 |
|---:|---|---:|---|
| `0` | Rahul | 21 | Python |
| `1` | Priya | 22 | Data Science |
| `2` | Aman | 20 | Machine Learning |

The first index selects a row, and the second index selects a column within that row.

## Common Beginner Mistakes

### Starting indexes at 1

Python indexing starts at `0`, not `1`.

### Reversing the indexes

The first index selects the student record. The second index selects information inside that record.

### Forgetting the second pair of brackets

Using only one index returns a complete inner list rather than one value.

### Using parentheses instead of square brackets

List indexes must be written inside square brackets.

### Adding a record with a different structure

Every new student record should follow the same order:

`[name, age, course]`

### Accidentally using a loop

This demonstration accesses each required value directly. No `for` or `while` loop is needed.
