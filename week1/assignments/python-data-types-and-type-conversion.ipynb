## 1. Numeric Data Types

Python provides different data types for working with numbers.

### Integer — `int`

An integer represents a whole number without a decimal point. Integers may be positive, negative, or zero.

Examples include age, year, number of students, and quantity.

### Float — `float`

A float represents a number containing a decimal point.

Examples include price, temperature, height, weight, and percentages.

We will create two integer variables and two float variables.

## 2. Strings and Boolean Values

### String — `str`

A string stores text. String values are written inside quotation marks.

Strings can contain letters, numbers, spaces, and special characters.

### Boolean — `bool`

A Boolean stores one of two possible values:

- `True`
- `False`

Boolean values are commonly used to represent yes-or-no conditions.

We will create two string variables and two Boolean variables.

## 3. Lists and Tuples

Lists and tuples allow us to store multiple values in one variable.

### List — `list`

A list is an ordered and changeable collection. List values are written inside square brackets.

A list can contain duplicate values.

### Tuple — `tuple`

A tuple is an ordered but unchangeable collection. Tuple values are normally written inside parentheses.

Like a list, a tuple can contain duplicate values.

We will create two lists and two tuples.

## 4. Sets and Dictionaries

### Set — `set`

A set stores unique values. Duplicate values are automatically removed.

Sets are unordered, so their displayed order should not be relied upon.

### Dictionary — `dict`

A dictionary stores information as key-value pairs.

Each key identifies a corresponding value. Dictionaries are useful for representing structured information such as profiles, products, and courses.

We will create one set and two dictionaries.

## Understanding Type Conversion

Type conversion means converting a value from one data type to another.

Python provides built-in conversion functions such as:

- `int()`
- `float()`
- `str()`
- `bool()`
- `list()`
- `tuple()`
- `set()`

For every conversion, we will display:

1. The original value
2. The original data type

## 6. Convert a String to an Integer

We will convert the string `"100"` into an integer using `int()`.

Before conversion, the value is text because it is inside quotation marks. After conversion, it becomes a whole number that can be used in mathematical calculations.

## 7. Convert a String to a Float

We will convert the string `"45.67"` into a floating-point number using `float()`.

Before conversion, the value is text. After conversion, it becomes a numeric decimal value that can be used in calculations.

## 8. Convert an Integer to a String

We will convert the integer `500` into a string using `str()`.

Before conversion, the value is a number. After conversion, it becomes text.

The displayed value may look similar, but its data type and available operations are different.

## 9. Convert an Integer to a Boolean

We will convert the integer `1` into a Boolean value using `bool()`.

In Python:

- `0` converts to `False`.
- Most nonzero numbers convert to `True`.

Therefore, converting `1` produces a Boolean result.

## 10. Convert a Tuple to a List

We will convert the tuple `(1, 2, 3)` into a list using `list()`.

The elements remain the same, but the collection changes from an immutable tuple to a mutable list.

A mutable collection can be changed after it is created.

## 11. Convert a List to a Tuple

We will convert the list `[1, 2, 3]` into a tuple using `tuple()`.

The elements remain the same, but the collection changes from a mutable list to an immutable tuple.

## 12. Convert a List to a Set

We will convert the list `[1, 2, 2, 3]` into a set using `set()`.

The original list contains the value `2` twice. Sets store only unique values, so the duplicate is removed during conversion.

The resulting set is unordered, which means its displayed order is not guaranteed.

## Type Conversion Summary

| Conversion | Original Type | Resulting Type | What Changes? |
|---|---|---|---|
| `int("100")` | String | Integer | Numeric text becomes a whole number |
| `float("45.67")` | String | Float | Decimal text becomes a numeric decimal |
| `str(500)` | Integer | String | A number becomes text |
| `bool(1)` | Integer | Boolean | A nonzero number becomes `True` |
| `list((1, 2, 3))` | Tuple | List | An immutable collection becomes mutable |
| `tuple([1, 2, 3])` | List | Tuple | A mutable collection becomes immutable |
| `set([1, 2, 2, 3])` | List | Set | Duplicate values are removed |

## Important Differences Between Collection Types

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Ordered | Yes | Yes | No guaranteed order | Yes |
| Changeable | Yes | No | Yes | Yes |
| Allows duplicates | Yes | Yes | No | Keys must be unique |
| Main syntax | `[ ]` | `( )` | `{ }` | `{key: value}` |

## Common Type-Conversion Mistakes

### Converting non-numeric text to a number

Python can convert `"100"` to an integer, but it cannot convert text such as `"hello"` to an integer.

### Converting decimal text directly to an integer

The string `"45.67"` is not formatted as a whole number. It should first be converted using `float()`.

### Assuming `str(500)` is still numeric

After conversion, `"500"` is text. Mathematical addition will not work in the same way as it does with the integer `500`.

### Expecting a set to retain duplicates

Sets contain only unique values, so duplicate elements are removed.

### Confusing `{}` with an empty set

Empty curly braces create an empty dictionary. An empty set is created using `set()`.
