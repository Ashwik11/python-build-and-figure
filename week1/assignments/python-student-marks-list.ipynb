## Method summary

| Function or method | Purpose | Changes the list? |
|---|---|---|
| `len(marks)` | Counts the elements | No |
| `max(marks)` | Finds the highest mark | No |
| `min(marks)` | Finds the lowest mark | No |
| `sum(marks)` | Adds all marks | No |
| `marks.sort()` | Sorts the list | Yes |
| `marks.append(95)` | Adds one element | Yes |
| `marks.extend([81, 84])` | Adds multiple elements | Yes |
| `marks.remove(67)` | Removes the first matching value | Yes |
| `marks.count(90)` | Counts matching values | No |
| `marks.index(88)` | Finds the first matching position | No |
