## Methods That Return Different Types

Although all these methods operate on strings, they do not all return strings:

- `strip()`, `lower()`, `upper()`, `title()`, and `replace()` return new strings.
- `startswith()` and `endswith()` return Boolean values.
- `find()` and `count()` return integers.
- `split()` returns a list of strings.

## Python Strings Are Immutable

Python strings are immutable, which means their existing characters cannot be changed directly.

String methods such as `strip()`, `lower()`, and `replace()` return new values. If we want to preserve a transformed value, we must store it in a variable or use it immediately.

## Common Mistakes to Avoid

### Using the original string for boundary checks

The original message contains surrounding spaces. Therefore, checking whether it starts with `Welcome` or ends with `Class` before using `strip()` may produce an unexpected result.

### Ignoring capitalization

Methods such as `replace()`, `count()`, `find()`, `startswith()`, and `endswith()` are case-sensitive.

### Expecting the original string to change

String methods return new strings. They do not automatically modify the value stored in the original variable.

### Confusing `find()` and `count()`

- `find()` reports where the first match begins.
- `count()` reports how many matches exist.
