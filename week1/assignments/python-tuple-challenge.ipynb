## Why Tuples Are Called Immutable

A tuple is called **immutable** because its elements cannot be added, removed, replaced, or rearranged after the tuple is created.

For example, assigning a new value to `technologies[0]` would raise a `TypeError`. We do not execute that invalid statement because it would stop the notebook.

To create an updated version, we followed this safe process:

1. Convert the tuple into a list.
2. Modify the list.
3. Convert the modified list back into a new tuple.

Immutability is useful when a collection should remain stable and protected from accidental modification.

## Tuple vs. List

| Feature | Tuple | List |
|---|---|---|
| Syntax | Parentheses `( )` | Square brackets `[ ]` |
| Ordered | Yes | Yes |
| Allows duplicates | Yes | Yes |
| Supports indexing and slicing | Yes | Yes |
| Changeable after creation | No | Yes |
| Common use | Fixed collections | Frequently changing collections |

## Complete Verification

The following independent cell performs every required operation in order.

```python
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")

print("Tuple:", technologies)
print("Type:", type(technologies))
print("First item:", technologies[0])
print("Last item:", technologies[-1])
print("Slice [1:4]:", technologies[1:4])
print("Python count:", technologies.count("Python"))
print("Index of C++:", technologies.index("C++"))
print("Tuple length:", len(technologies))

technology_list = list(technologies)
print("Converted list:", technology_list)

technology_list.append("Go")
print("List after adding Go:", technology_list)

updated_technologies = tuple(technology_list)
print("Converted back to tuple:", updated_technologies)
```
