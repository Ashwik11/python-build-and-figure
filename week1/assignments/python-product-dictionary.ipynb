## Dictionary Method Summary

| Operation | Purpose |
|---|---|
| `dictionary["key"]` | Accesses a value |
| `dictionary["key"] = value` | Adds a new key or updates an existing key |
| `pop("key")` | Removes a key and returns its value |
| `keys()` | Returns all keys |
| `values()` | Returns all values |
| `items()` | Returns all key-value pairs |

## Complete Verification

The following independent cell re-creates the original laptop and performs every required operation in order.

```python
laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}

print("Product:", laptop)
print("Brand:", laptop["brand"])
print("Model:", laptop["model"])
print("Price:", laptop["price"])

laptop["price"] = 125000
laptop["processor"] = "Intel Core i7"
laptop["gpu"] = "NVIDIA RTX 4060"
laptop["ram"] = "32GB"
removed_availability = laptop.pop("available")

print("Updated price:", laptop["price"])
print("Processor:", laptop["processor"])
print("GPU:", laptop["gpu"])
print("Updated RAM:", laptop["ram"])
print("Removed availability:", removed_availability)
print("All keys:", laptop.keys())
print("All values:", laptop.values())
print("All items:", laptop.items())
print("Final laptop:", laptop)

mobile_phone = {
    "brand": "Samsung",
    "model": "Galaxy S25",
    "price": 90000,
    "ram": "12GB",
    "storage": "256GB",
    "operating_system": "Android",
    "available": True
}
print("Mobile phone:", mobile_phone)
```
