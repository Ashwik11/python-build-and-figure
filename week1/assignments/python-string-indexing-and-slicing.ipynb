## Understanding String Indexing

Every character in a Python string has a numerical position called an index.

Python uses zero-based indexing:

- The first character is at index `0`.
- The second character is at index `1`.
- The third character is at index `2`.

Python also supports negative indexing:

- The last character is at index `-1`.
- The second-to-last character is at index `-2`.

Spaces are also characters, so they have index positions.

## Understanding String Slicing

String slicing allows us to extract a portion of a string.

The general syntax is:

`string[start:stop]`

- `start` is the index where the slice begins.
- `stop` is the index where the slice ends.
- The character at the `stop` index is not included.

This is sometimes described as “start inclusive, stop exclusive.”

## Indexing, Slicing, and Method Summary

| Operation | Purpose |
|---|---|
| `string[0]` | Accesses the first character |
| `string[-1]` | Accesses the last character |
| `string[:n]` | Extracts the first `n` characters |
| `string[-n:]` | Extracts the last `n` characters |
| `string[::-1]` | Reverses the string |
| `upper()` | Converts text to uppercase |
| `lower()` | Converts text to lowercase |
| `title()` | Converts text to title case |
| `count()` | Counts matching characters or substrings |
| `find()` | Returns the starting index of a match |
| `replace()` | Replaces selected text |
| `split()` | Divides a string into a list |

## Important Observations

- Python string indexes begin at `0`.
- Negative indexes count backward from the end.
- The stopping index in a slice is not included.
- Spaces are included as characters in a string.
- String methods are case-sensitive when searching for specific text.
- Python strings are immutable, so these operations return new values.

## Common Beginner Mistakes

### Starting with index 1

The first character is at index `0`, not index `1`.

### Including the stopping index

In a slice such as `string[0:6]`, index `6` is not included.

### Forgetting that spaces have indexes

Spaces affect the positions returned by methods such as `find()`.

### Using the wrong capitalization

Searching for `Programming` will produce a different result from searching for `programming`.

### Expecting the original string to change

Methods such as `upper()` and `replace()` return new strings. They do not automatically modify the original variable.
