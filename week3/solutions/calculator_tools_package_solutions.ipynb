# Build Your Own `calculator_tools` Python Package — Solutions


This notebook demonstrates a reusable Python package containing arithmetic, percentage, average, temperature-conversion, unit-conversion, validation, and custom-exception functionality. All solution files are stored in the required `calculator_tools/` folder.

## Function, module, package, and import

| Concept | Meaning | Example in this project |
|---|---|---|
| Function | A reusable block of code that performs one task | `add()`, `average()`, `convert_unit()` |
| Module | One Python `.py` file | `arithmetic.py` |
| Package | A folder of related modules containing `__init__.py` | `calculator_tools/` |
| Import | A statement that makes code from another module or package available | `from calculator_tools import add` |

`__init__.py` defines the package's public interface, so users can import common functions directly from `calculator_tools`.

## Required project structure

```text
calculator_tools/
├── __init__.py
├── arithmetic.py
├── statistics.py
├── converter.py
├── exceptions.py
└── main.py
```

Run the demonstration from the project root with:

```bash
python -m calculator_tools.main
```

## 3. Import the package

These imports use the public interface defined in `calculator_tools/__init__.py`.

````python
from calculator_tools import (
    InvalidOperationError,
    add,
    average,
    calculate,
    calculate_average,
    calculate_percentage,
    celsius_to_fahrenheit,
    convert_temperature,
    convert_unit,
    divide,
    fahrenheit_to_celsius,
    kilograms_to_pounds,
    kilometers_to_miles,
    multiply,
    subtract,
)

print("calculator_tools imported successfully.")
````

---

## 4. Basic arithmetic and percentage

Each operation is a separate reusable function. The optional `calculate()` dispatcher selects a function from an operation name or symbol.

````python
print("Addition:", add(15, 5))
print("Subtraction:", subtract(15, 5))
print("Multiplication:", multiply(15, 5))
print("Division:", divide(15, 5))
print("Dispatcher:", calculate(9, 4, "multiply"))
print("Percentage:", calculate_percentage(45, 60), "%")
````

---

## 5. Average calculation

`average()` requires a non-empty iterable in which every item is numeric. `calculate_average()` is a readable alternative name for the same behavior.

````python
marks = [80, 90, 70, 100]
print("Average marks:", average(marks))
print("Average of a tuple:", calculate_average((10, 20, 30)))
````

---

## 6. Temperature conversion

The package supports Celsius-to-Fahrenheit and Fahrenheit-to-Celsius conversions. Values below absolute zero are rejected.

````python
print("25 C in Fahrenheit:", celsius_to_fahrenheit(25))
print("77 F in Celsius:", fahrenheit_to_celsius(77))
print("0 C in Fahrenheit using dispatcher:", convert_temperature(0, "C", "F"))
````

---

## 7. Simple unit conversion

Distance conversions support kilometers and miles. Weight conversions support kilograms and pounds. Negative measurements are rejected.

````python
print("10 km in miles:", round(kilometers_to_miles(10), 3))
print("5 kg in pounds:", round(kilograms_to_pounds(5), 3))
print("10 miles in km:", round(convert_unit(10, "miles", "km"), 3))
print("22 pounds in kg:", round(convert_unit(22, "lb", "kg"), 3))
````

---

## 8. Error handling and custom exception

The examples below deliberately provide invalid inputs. Each exception is caught so the notebook continues running. Unsupported operations raise the custom `InvalidOperationError`.

````python
error_examples = [
    ("Division by zero", lambda: divide(8, 0)),
    ("Incorrect data type", lambda: add("8", 2)),
    ("Empty average", lambda: average([])),
    ("Non-numeric list item", lambda: average([10, "twenty"])),
    ("Temperature below absolute zero", lambda: celsius_to_fahrenheit(-300)),
    ("Negative unit value", lambda: kilometers_to_miles(-1)),
    ("Unsupported arithmetic operation", lambda: calculate(2, 3, "power")),
    ("Unsupported unit conversion", lambda: convert_unit(1, "meter", "liter")),
]

for label, action in error_examples:
    try:
        action()
    except (InvalidOperationError, TypeError, ValueError, ZeroDivisionError) as error:
        print(f"{label}: {type(error).__name__} - {error}")
````

## 10. Automated solution checks

Assertions provide a quick final check that the important results and custom exception behave correctly.

````python
assert add(2, 3) == 5
assert subtract(10, 4) == 6
assert multiply(6, 7) == 42
assert divide(9, 3) == 3
assert calculate_percentage(45, 60) == 75
assert average([2, 4, 6]) == 4
assert celsius_to_fahrenheit(0) == 32
assert fahrenheit_to_celsius(32) == 0
assert round(kilometers_to_miles(1), 6) == 0.621371

try:
    calculate(2, 3, "power")
except InvalidOperationError:
    pass
else:
    raise AssertionError("Unsupported operation did not raise InvalidOperationError.")

print("All automated checks passed.")
````
