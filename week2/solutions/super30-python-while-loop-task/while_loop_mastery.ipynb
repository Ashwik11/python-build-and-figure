## Question 1 — Print numbers from 1 to 100

### Problem
Print every integer from 1 through 100 using a `while` loop.

### How to approach it

- Initialize a counter at the first value to print.
- Continue while the counter remains within the required upper boundary.
- Print the current value during each iteration.
- Increase the counter so the loop eventually ends.

### Expected result
The output should begin with `1`, end with `100`, and contain 100 values.



````python
counter = 1
while counter <= 100:
    print(counter, end = " ")
    counter += 1
````

## Question 4 — Sum the digits of an integer

### Problem
Calculate the sum of all digits in an integer.

**Example:** `5832` → `18`

### How to approach it

- Preserve the original input if you want to display it later.
- For a negative input, work with its absolute value so the sign is not treated as a digit.
- Initialize a digit-total accumulator.
- Repeatedly isolate the final digit with the remainder operation.
- Add that digit to the accumulator.
- Remove the final digit using integer division.
- Stop when no digits remain.

### Suggested tests

- `5832` → `18`
- `1005` → `6`


````python
n = int(input())
sum = 0
n = abs(n)
while n > 0 :
    rem = n % 10
    sum += rem
    n = n // 10
print(sum)
````

## Question 5 — Reverse an integer

### Problem
Reverse an integer using a `while` loop.

**Example:** `12345` → `54321`

### How to approach it

- Use a working copy of the input.
- Initialize the reversed result at zero.
- Repeatedly extract the final digit of the working number.
- Shift the current reversed result one decimal place and attach the extracted digit.
- Remove the extracted digit from the working number.
- Stop when no digits remain.

### Suggested tests

- `12345` → `54321`
- `7` → `7`
- `1200` → `21` because an integer does not preserve leading zeros



````python
n = int(input())
copy = abs(n)
reversed_number = 0
while(copy > 0):
    digit = copy % 10
    reversed_number = reversed_number * 10 + digit
    copy //= 10

print(reversed_number)

````

## Question 6 — Count the digits in an integer

### Problem
Count how many digits an integer contains using a `while` loop.

### How to approach it

- Use the absolute value if negative inputs are supported.
- Initialize a digit counter.
- Handle zero explicitly because `0` contains one digit even though a “while value is nonzero” loop would not run.
- For other values, repeatedly remove the final digit and increase the counter.

### Suggested tests

- `5832` → `4`
- `7` → `1`
- `0` → `1`
- `-2048` → `4`


````python
n = int(input())
copy = abs(n)
counter = 0
if copy == 0:
    counter = 1
else:
    while(copy > 0):
        copy //= 10
        counter += 1
print(counter)
````

Calculate the factorial of a non-negative integer using a `while` loop.

### How to approach it

- Reject or clearly handle negative input because factorial is not defined for negative integers in this assignment.
- Initialize the product using the multiplicative identity.
- Initialize a counter for the first factor.

### Suggested tests

- `5` → `120`
- `1` → `1`
- `0` → `1`

````python
n = int(input())
if n < 0 :
    print("factortila not defined for negative numbers")
else:
    factorial = 1
    while(n > 0):
        factorial = n * factorial
        n -= 1
    print(factorial)
````

## Question 8 — Sum inputs until the user enters 0

### Problem
Repeatedly ask the user for numbers. Stop only when the user enters `0`, then display the sum of all previously entered numbers.

### How to approach it

- Initialize the running total before the loop.
- Repeatedly request and convert a number.
- Treat zero as a sentinel: it controls termination and must not change the total.
- Add every nonzero entry to the running total.
- Print the total after the loop ends.

### Suggested tests

- Inputs `10`, `-3`, `5`, `0` → total `12`
- First input `0` → total `0`

### Why `while` is preferable here
The number of entries is unknown. The loop must continue until a value, rather than a predetermined count, signals completion.

````python
total = 0 
while(True):
    n = int(input())
    if n != 0:
        total += n
    else:
        break
print(total)
````

## Question 12 — ATM menu using `while`

### Required behavior
The ATM should continue running until the user explicitly selects Exit and should allow the user to:

1. Check balance
2. Deposit money
3. Withdraw money
4. Exit

### How to approach it

- Initialize a balance before the menu loop.
- Display and process the menu repeatedly.
- Accept only positive deposit and withdrawal amounts.
- Reject withdrawals that exceed the available balance.
- Update the balance only after a valid transaction.
- Handle invalid menu choices without ending the application.
- Use clear branches or small helper functions instead of one unreadable block.
- End the loop only through the explicit Exit choice.

### Required demonstration

- Check the starting balance.
- Complete a valid deposit.
- Complete a valid withdrawal.
- Attempt an overdraw.
- Attempt a zero or negative transaction.
- Enter an invalid menu choice.
- Exit and confirm that the menu stops.

### Why `while` is preferable here
An ATM session has no predetermined number of transactions. Its natural termination condition is the user's Exit choice.



````python
balance = 10000

while True:
    print("\n1. Check balance")
    print("2. Deposit money")
    print("3. Withdraw money")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        print(f"Current balance: {balance}")

    elif choice == "2":
        amount = float(input("Enter deposit amount: "))

        if amount > 0:
            balance += amount
            print(f"Deposit successful. New balance: {balance}")
        else:
            print("Deposit amount must be positive.")

    elif choice == "3":
        amount = float(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrawal successful. New balance: {balance}")

    elif choice == "4":
        print("Thank you for using the ATM.")
        break

    else:
        print("Invalid option. Please choose 1, 2, 3, or 4.")
````
