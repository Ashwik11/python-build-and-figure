## Core vocabulary for your video

- **Function definition:** The `def` statement that creates a reusable block of behavior.
- **Function call:** The expression that runs the function.
- **Parameter:** A variable named in the function definition.
- **Argument:** A value supplied when calling the function.
- **Return value:** A result sent back to the caller with `return`.
- **Default argument:** A parameter value used when the caller does not provide one.
- **Positional argument:** An argument matched according to its position.
- **Keyword argument:** An argument matched by explicitly naming its parameter.
- **Local scope:** Variables created inside a function normally exist only during that function call.

> **Required rule:** Every custom function must contain a concise docstring immediately below its definition.
