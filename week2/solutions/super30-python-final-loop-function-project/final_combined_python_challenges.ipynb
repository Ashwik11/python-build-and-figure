# Final Combined Challenge: Loops, `while`, and Functions

Build 12 practical Python mini-programs that combine reusable functions with count-controlled and condition-controlled iteration.

Each project below includes design guidance, edge cases, and an empty implementation cell. The programs are intentionally independent so they can later be placed in separate Python files.

## Design principles

- Give each function one clear responsibility.
- Put problem-solving logic inside functions rather than calculating globally.
- Add concise docstrings to custom functions.
- Use `for` when iterating through a known collection or range.
- Use `while` when repetition depends on a changing condition or an Exit choice.
- Validate user input before changing stored data.
- Use meaningful names and keep menu branches small.
- Return reusable values from calculation functions; display them at the application boundary.
- Test successful, unsuccessful, boundary, and empty-data cases.

> The strongest implementation is not the one with the fewest lines—it is the one whose behavior is easy to understand, test, and extend.

## Suggested workflow

For each mini-project:

1. Decide how the data will be represented.
2. List the required functions and what each one returns.
3. Identify where a `for` loop or `while` loop is appropriate.
4. Implement the smallest useful function first.
5. Connect the functions through a controller or `main()` function.
6. Run the suggested test flow and inspect every state change.
7. Keep the final cell free of unhandled errors.

## Project 1 — Student Result Management System

### Goal
Create a program that accepts student marks, calculates the result, and displays a clear summary.

### Suggested function responsibilities

- Accept and validate marks.
- Calculate the total.
- Calculate the percentage.
- Assign a grade.
- Determine pass or fail.
- Format and display the final result.

### Design guidance

- Store the marks in a list.
- Use a loop to collect or process subject marks.
- Validate that every mark falls within the accepted range.
- Decide and document the grading thresholds and pass rule before coding because they are not specified.
- Return values from calculation functions so the display function can combine them.
- Keep input, calculation, and presentation responsibilities separate.

### Suggested test flow

- A student who passes every subject
- A student with a high overall grade
- A student who fails at least one subject
- A mark at every important grade boundary
- An invalid mark below 0 or above 100

### Design check
Does changing the grading scale require editing only the grade function, or does the rule appear in several places?

```python
def validate_marks(marks):
    """Return marks when every value is between 0 and 100."""
    if any(mark < 0 or mark > 100 for mark in marks):
        raise ValueError("Marks must be between 0 and 100")
    return marks

def total_marks(marks):
    """Return the total of the marks."""
    total = 0
    for mark in marks:
        total += mark
    return total

def grade_for(percentage):
    """Return a grade for a percentage."""
    if percentage >= 90: return "A"
    if percentage >= 75: return "B"
    if percentage >= 60: return "C"
    if percentage >= 40: return "D"
    return "F"

def student_result(marks):
    """Return a student's result summary."""
    validate_marks(marks)
    total = total_marks(marks)
    percentage = total / len(marks)
    passed = all(mark >= 40 for mark in marks)
    return {"total": total, "percentage": percentage, "grade": grade_for(percentage), "result": "Pass" if passed else "Fail"}

print(student_result([85, 92, 78]))
```

## Project 2 — Banking Application

### Goal
Build a menu-driven banking program with balance checking, deposits, withdrawals, transaction history, and Exit.

### Suggested function responsibilities

- Display the current balance.
- Validate and process a deposit.
- Validate and process a withdrawal.
- Display transaction history.
- Control the application menu.

### Design guidance

- Store the balance as numeric state.
- Store transaction records in a list; each record can include a type, amount, and resulting balance.
- Use a `while` loop because the user decides when the session ends.
- Accept only positive transaction amounts.
- Reject withdrawals that exceed the balance.
- Record only completed transactions.
- Return the updated balance rather than relying unnecessarily on a global variable.
- Provide a helpful response for an invalid menu choice.

### Suggested test flow

- Check the opening balance.
- Complete a deposit and withdrawal.
- Attempt an overdraw.
- Attempt a zero or negative amount.
- Inspect the transaction history.
- Enter an invalid choice, then exit cleanly.

### Design check
Can every failed operation leave both the balance and transaction history unchanged?

```python
def deposit(balance, history, amount):
    """Deposit a positive amount and record the transaction."""
    if amount <= 0:
        return balance, "Amount must be positive"
    balance += amount
    history.append(("Deposit", amount, balance))
    return balance, "Deposit successful"

def withdraw(balance, history, amount):
    """Withdraw an affordable positive amount and record it."""
    if amount <= 0:
        return balance, "Amount must be positive"
    if amount > balance:
        return balance, "Insufficient balance"
    balance -= amount
    history.append(("Withdrawal", amount, balance))
    return balance, "Withdrawal successful"

def main():
    """Run the banking menu."""
    balance = 1000.0
    history = []
    while True:
        print("1. Balance  2. Deposit  3. Withdraw  4. History  5. Exit")
        choice = input("Choose: ")
        if choice == "1": print(balance)
        elif choice == "2":
            balance, message = deposit(balance, history, float(input("Amount: ")))
            print(message)
        elif choice == "3":
            balance, message = withdraw(balance, history, float(input("Amount: ")))
            print(message)
        elif choice == "4": print(history)
        elif choice == "5": break
        else: print("Invalid choice")
```

## Project 3 — Inventory Management

### Goal
Maintain products with a name, price, and quantity, and support common inventory operations.

### Required capabilities

- Add a product.
- Display all products.
- Search for a product.
- Update a product's quantity.
- Calculate the total inventory value.

### Design guidance

- A dictionary keyed by a normalized product name makes searching and updating direct.
- Each product record can store price and quantity.
- Decide whether adding an existing product should be rejected or treated as an update.
- Validate that price is not negative and quantity is a valid non-negative integer.
- Calculate each product's inventory value as price multiplied by quantity, then accumulate the total.
- Handle searches and updates for missing products without crashing.
- Keep menu control separate from inventory operations if you add an interactive menu.

### Suggested test flow

- Add two valid products.
- Attempt a duplicate product.
- Search for an existing and missing product.
- Update quantity to a valid value and to zero.
- Reject negative price or quantity.
- Verify the total inventory value manually.

### Design check
Does product-name normalization prevent accidental duplicates such as `Mouse` and `mouse`?

```python
def add_product(inventory, name, price, quantity):
    """Add a new normalized product to inventory."""
    key = name.strip().lower()
    if not key or price < 0 or quantity < 0 or quantity != int(quantity):
        return False
    if key in inventory:
        return False
    inventory[key] = {"name": name.strip(), "price": price, "quantity": int(quantity)}
    return True

def search_product(inventory, name):
    """Return a product or None when it is missing."""
    return inventory.get(name.strip().lower())

def update_quantity(inventory, name, quantity):
    """Update a product quantity when the product and value are valid."""
    product = search_product(inventory, name)
    if product is None or quantity < 0 or quantity != int(quantity):
        return False
    product["quantity"] = int(quantity)
    return True

def inventory_value(inventory):
    """Return the total value of all products."""
    total = 0
    for product in inventory.values():
        total += product["price"] * product["quantity"]
    return total

inventory = {}
add_product(inventory, "Mouse", 700, 3)
add_product(inventory, "Keyboard", 1500, 2)
print(inventory)
print(search_product(inventory, "mouse"))
update_quantity(inventory, "Mouse", 0)
print(inventory_value(inventory))
```

## Project 5 — Number Analysis Tool

### Goal
Create a function that accepts a list and returns:

- Largest number
- Smallest number
- Total
- Average
- Even count
- Odd count
- Positive count
- Negative count

Do not use `min()`, `max()`, or `sum()`.

### Design guidance

- Decide how an empty list will be handled before reading its first element.
- Initialize largest and smallest from an actual list value.
- Initialize the total and four counters.
- Analyze all values in one loop.
- Treat zero as even, but neither positive nor negative.
- Calculate the average only after the total and count are known.
- Return all results in a structured form such as a dictionary.

### Suggested tests

- A mixture of positive, negative, even, odd, and zero values
- A list containing one value
- A list containing only negative numbers
- An empty list

### Design check
Can the list be traversed once while still producing every required statistic?

```python
def analyze_numbers(numbers):
    """Return statistics for a list without min, max, or sum."""
    if not numbers:
        return {"largest": None, "smallest": None, "total": 0, "average": None, "even": 0, "odd": 0, "positive": 0, "negative": 0}
    largest = smallest = numbers[0]
    total = even = odd = positive = negative = 0
    for number in numbers:
        total += number
        largest = number if number > largest else largest
        smallest = number if number < smallest else smallest
        if number % 2 == 0: even += 1
        else: odd += 1
        if number > 0: positive += 1
        elif number < 0: negative += 1
    return {"largest": largest, "smallest": smallest, "total": total, "average": total / len(numbers), "even": even, "odd": odd, "positive": positive, "negative": negative}

print(analyze_numbers([-3, 0, 4, 7, 10]))
```

## Project 6 — Employee Salary Analyzer

### Goal
Analyze employee salaries to determine total payroll, average salary, highest salary, lowest salary, and employees earning above average.

### Suggested data structure
Use a dictionary mapping employee names to salaries, or a list of employee records if more fields may be added later.

### Suggested function responsibilities

- Calculate total payroll.
- Calculate average salary.
- Find the highest and lowest salaries.
- Return the employees whose salaries exceed the average.
- Display the analysis.

### Design guidance

- Validate non-negative numeric salaries.
- Handle an empty collection before dividing or accessing the first record.
- Calculate the average before filtering above-average employees.
- Decide how to report ties for highest or lowest salary.
- Prefer returning names together with salaries for meaningful output.

### Suggested test flow

- Several different salaries
- Two employees tied for highest salary
- One employee exactly at the average
- A single employee
- No employees

### Design check
Does “above average” use a strict comparison so an employee exactly at the average is excluded?

```python
def salary_analysis(salaries):
    """Return payroll statistics and employees above average."""
    if not salaries:
        return {"total": 0, "average": None, "highest": [], "lowest": [], "above_average": []}
    total = 0
    highest = lowest = None
    for salary in salaries.values():
        if salary < 0:
            raise ValueError("Salary cannot be negative")
        total += salary
        highest = salary if highest is None or salary > highest else highest
        lowest = salary if lowest is None or salary < lowest else lowest
    average = total / len(salaries)
    return {"total": total, "average": average, "highest": [(name, salary) for name, salary in salaries.items() if salary == highest], "lowest": [(name, salary) for name, salary in salaries.items() if salary == lowest], "above_average": [(name, salary) for name, salary in salaries.items() if salary > average]}

print(salary_analysis({"Asha": 50000, "Ben": 70000, "Cara": 70000, "Dev": 60000}))
```

## Project 7 — Shopping Cart

### Goal
Create a menu-driven cart that lets users add products, remove products, view the cart, calculate the bill, and exit.

### Suggested data structure
Represent the cart with product names linked to price and quantity. A separate product catalog can prevent users from inventing prices.

### Suggested function responsibilities

- Add a product and quantity.
- Remove a product or reduce its quantity.
- Display cart contents.
- Calculate the total bill.
- Control the menu.

### Design guidance

- Use a `while` loop because the user controls when shopping ends.
- Validate positive prices and quantities.
- Decide how adding an existing product changes its quantity.
- Handle removal from an empty cart and requests for missing products.
- Remove an item completely when its quantity reaches zero.
- Calculate the bill from current price and quantity values.

### Suggested test flow

- Add two products and add one of them twice.
- View the combined quantity.
- Remove part of a quantity, then remove the rest.
- Attempt to remove a missing product.
- Calculate an empty and non-empty bill.
- Exit cleanly.

### Design check
Is one product represented once with a quantity, rather than duplicated as several unrelated cart entries?

```python
def add_to_cart(cart, name, price, quantity):
    """Add a product or increase its quantity."""
    if price <= 0 or quantity <= 0:
        return False
    item = cart.setdefault(name, {"price": price, "quantity": 0})
    item["quantity"] += quantity
    return True

def remove_from_cart(cart, name, quantity):
    """Reduce a product quantity and remove it at zero."""
    if name not in cart or quantity <= 0:
        return False
    cart[name]["quantity"] -= quantity
    if cart[name]["quantity"] <= 0:
        del cart[name]
    return True

def cart_total(cart):
    """Return the current cart total."""
    total = 0
    for item in cart.values():
        total += item["price"] * item["quantity"]
    return total

cart = {}
add_to_cart(cart, "Book", 20, 2)
add_to_cart(cart, "Book", 20, 1)
print(cart, cart_total(cart))
remove_from_cart(cart, "Book", 1)
print(cart, cart_total(cart))
```

## Project 8 — Password Strength Checker

### Goal
Create a function that checks whether a password contains:

- At least one uppercase letter
- At least one lowercase letter
- At least one number
- At least one special character
- At least eight characters

Return a meaningful result.

### Design guidance

- Initialize one Boolean flag for each character requirement.
- Iterate through the password once and update the flags.
- Treat a special character as a character that is not a letter or digit, or use a clearly defined punctuation set.
- Check the length requirement separately.
- Return a structured result, a list of missing requirements, or a descriptive strength message.
- Do not print the password itself in diagnostic output.

### Suggested tests

- A password satisfying all five requirements
- A password shorter than eight characters
- Missing uppercase, lowercase, number, and special-character cases
- An empty string

### Design check
Can the caller tell exactly which requirements failed instead of receiving only `Weak`?

```python
def password_strength(password):
    """Return password requirements that are still missing."""
    has_upper = has_lower = has_number = has_special = False
    for character in password:
        if character.isupper(): has_upper = True
        elif character.islower(): has_lower = True
        elif character.isdigit(): has_number = True
        else: has_special = True
    missing = []
    if not has_upper: missing.append("uppercase letter")
    if not has_lower: missing.append("lowercase letter")
    if not has_number: missing.append("number")
    if not has_special: missing.append("special character")
    if len(password) < 8: missing.append("at least 8 characters")
    return {"strong": not missing, "missing": missing}

print(password_strength("Python@123"))
print(password_strength("hello"))
```

## Project 9 — Prime Number Analyzer

### Goal
Accept two numbers defining a range, then find all primes, count them, calculate their sum, and display the largest prime found.

### Suggested function responsibilities

- Determine whether one number is prime.
- Find all primes in an inclusive range.
- Count the prime list.
- Calculate the prime total.
- Find the largest discovered prime.
- Display the analysis.

### Design guidance

- Treat values below 2 as non-prime.
- Normalize reversed boundaries or reject them with a clear message.
- Reuse the single-number prime checker instead of duplicating divisor logic.
- Avoid `sum()` and `max()` if you want to reinforce manual loop analysis.
- Handle a range containing no primes before requesting the largest prime.

### Suggested tests

- Range 1 through 10 → primes `2, 3, 5, 7`, count `4`, total `17`, largest `7`
- A single-value prime range
- A range with no primes
- Reversed boundaries
- Negative boundaries

### Design check
Is primality logic written once and reused by the range function?

```python
def is_prime(number):
    """Return whether a number is prime."""
    if number < 2:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True

def prime_analysis(start, end):
    """Return prime values and their summary for an inclusive range."""
    if start > end:
        start, end = end, start
    primes = []
    for number in range(start, end + 1):
        if is_prime(number): primes.append(number)
    total = 0
    for number in primes: total += number
    return {"primes": primes, "count": len(primes), "total": total, "largest": primes[-1] if primes else None}

print(prime_analysis(1, 10))
print(prime_analysis(10, 1))
```

## Project 10 — Expense Tracker

### Suggested function responsibilities

- Add a validated expense.

### Design guidance

- Use a `while` loop for the session menu.
- Reject empty names and non-positive amounts.
- Number expenses when displaying them.
- Handle total and highest calculations when the list is empty.
- Return calculated values rather than depending on hidden global state.

### Suggested test flow

- Add two expenses with the same name.
- Attempt an empty name and zero or negative amount.
- Verify the total manually.
- Find the highest expense.
- Test calculations before any expense exists.

### Design check
Can repeated real-world expenses such as two separate coffee purchases both be retained?

```python
def add_expense(expenses, name, amount):
    """Add a named expense when its values are valid."""
    if not name.strip() or amount <= 0:
        return False
    expenses.append({"name": name.strip(), "amount": amount})
    return True

def expense_total(expenses):
    """Return total spending."""
    total = 0
    for expense in expenses: total += expense["amount"]
    return total

def highest_expense(expenses):
    """Return the highest expense or None for an empty list."""
    if not expenses: return None
    highest = expenses[0]
    for expense in expenses[1:]:
        if expense["amount"] > highest["amount"]: highest = expense
    return highest

def show_expenses(expenses):
    """Display numbered expenses."""
    for index, expense in enumerate(expenses, 1): print(index, expense["name"], expense["amount"])

expenses = []
add_expense(expenses, "Coffee", 4.5)
add_expense(expenses, "Coffee", 5)
show_expenses(expenses)
print(expense_total(expenses), highest_expense(expenses))
```

## Project 11 — Mini Authentication System

### Goal
Create a small application with predefined credentials, limited login attempts, success and failure handling, logout, and retry logic.

### Suggested function responsibilities

- Validate one username-password pair.
- Run a login sequence with a maximum attempt count.
- Run the authenticated session.
- Log out and return to the appropriate state.
- Control whether another login sequence begins.

### Design guidance

- Store only the predefined test credentials needed for this learning project.
- Track attempts explicitly and never exceed the configured maximum.
- Count a failed credential pair as one attempt.
- Start the authenticated menu only after a successful login.
- Ensure logout changes the authentication state before retrying.
- Decide whether exhausting attempts ends the application or offers a clearly controlled restart.
- Avoid displaying the expected password in failure messages.

### Suggested test flow

- Correct credentials on the first attempt
- Incorrect password followed by success
- Incorrect username and password until attempts are exhausted
- Successful login followed by logout and another login

### Design check
Are the logged-out, authenticating, authenticated, and locked-or-ended states clearly separated?

```python
def valid_login(username, password):
    """Return whether the credentials match the stored account."""
    return username == "admin" and password == "python123"

def login(max_attempts=3):
    """Allow a limited number of login attempts."""
    for attempt in range(max_attempts):
        username = input("Username: ")
        password = input("Password: ")
        if valid_login(username, password):
            return True
        print("Invalid credentials")
    return False

def session():
    """Run the authenticated session until logout."""
    while True:
        choice = input("1. Profile  2. Logout: ")
        if choice == "1": print("Welcome, admin")
        elif choice == "2": return
        else: print("Invalid choice")

def main():
    """Control authentication and logout."""
    while login():
        session()
        if input("Try again? (y/n): ").lower() != "y": break

# main()
```

## Project 12 — Python Utility Application

### Goal
Create one menu-driven application containing at least five reusable utilities.

### Possible utilities

- Calculator
- Palindrome checker
- Prime checker
- Factorial calculator
- Multiplication table generator
- Number analyzer
- Password checker
- Unit converter
- Text statistics

### Recommended structure

- Implement every utility as its own function or small group of focused functions.
- Create one menu that maps choices to utilities.
- Use a `while` loop so the application continues until Exit.
- Keep the menu controller responsible for navigation, not individual calculations.
- Validate input within the appropriate utility.
- Return to the main menu after each completed or rejected operation.
- Provide an invalid-choice response without terminating the application.
- Add your own feature only if you can describe and test every line.

### Suggested test flow

- Run all five selected utilities successfully.
- Test one important failure or boundary case for each utility.
- Enter an invalid main-menu choice.
- Confirm that each utility returns to the menu.
- Exit cleanly.

### Design check
Can a sixth utility be added with a new function and one menu route without rewriting the existing utilities?

```python
def calculator(a, operator, b):
    """Return a calculator result or an error message."""
    if operator == "+": return a + b
    if operator == "-": return a - b
    if operator == "*": return a * b
    if operator == "/":
        return "Cannot divide by zero" if b == 0 else a / b
    return "Invalid operator"

def palindrome(text):
    """Return whether text reads the same forwards and backwards."""
    return text == text[::-1]

def is_prime(number):
    """Return whether a number is prime."""
    if number < 2: return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0: return False
    return True

def factorial(number):
    """Return a non-negative integer's factorial."""
    if number < 0: return "Number must be non-negative"
    result = 1
    for value in range(2, number + 1): result *= value
    return result

def utility_main():
    """Run a menu for reusable utilities."""
    utilities = {"1": "Calculator", "2": "Palindrome", "3": "Prime", "4": "Factorial", "5": "Exit"}
    while True:
        print(utilities)
        choice = input("Choose: ")
        if choice == "5": break
        if choice == "1": print(calculator(float(input()), input("Operator: "), float(input())))
        elif choice == "2": print(palindrome(input("Text: ")))
        elif choice == "3": print(is_prime(int(input("Number: "))))
        elif choice == "4": print(factorial(int(input("Number: "))))
        else: print("Invalid choice")

# utility_main()
```
