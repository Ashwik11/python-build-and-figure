## Challenge 1 — Total any number of values with `*args`

### Goal
Create a function that accepts any number of positional values and returns their total.

### Approach

- Define one function with an `*args` parameter.
- Recognize that the supplied values are available inside the function as a tuple.
- Initialize a local accumulator and process every value.
- Return the final total.
- Decide what an empty call should return; zero is a natural result.

### Suggested tests

- Three positive values
- A mixture of positive and negative values
- One value
- No values

### Concept check
What is the type of the `args` variable inside the function, and why can the caller supply different numbers of arguments?

````python
def total(*args):
    """Return the total of all supplied values."""
    result = 0
    for value in args:
        result += value
    return result

print(total(10, 20, 30))
print(total(-5, 2, 3))
print(total())
````

## Challenge 2 — Largest supplied value with `*args`

### Goal
Create a function that accepts any number of positional values and returns the largest supplied number.

### Approach

- Receive the values through `*args`.
- Handle an empty call explicitly because no largest value exists.
- Initialize the tracked largest value from an actual supplied argument.
- Compare the remaining values and update the tracked result when necessary.
- Return the result after all comparisons.

### Suggested tests

- `(10, 4, 27, 9)` → `27`
- `(-8, -2, -15)` → `-2`
- `(5,)` → `5`
- No arguments → a clear error or validation result

### Concept check
Why is initializing the largest value to zero unsafe when all supplied values are negative?

````python
def largest(*args):
    """Return the largest supplied value."""
    if not args:
        raise ValueError("At least one value is required")
    result = args[0]
    for value in args[1:]:
        if value > result:
            result = value
    return result

print(largest(10, 4, 27, 9))
print(largest(-8, -2, -15))
````

## Challenge 3 — Dynamic profile with `**kwargs`

### Required signature

```python
def create_profile(**kwargs):
```

### Goal
Accept dynamic user information and print every provided attribute.

### Approach

- Recognize that `kwargs` becomes a dictionary inside the function.
- Iterate through its key-value pairs.
- Display each attribute in a consistent format.
- Call the function with different sets of named information to demonstrate flexibility.
- Decide how an empty call should behave.

### Suggested calls

- A profile containing name and age
- A profile containing name, course, city, and profession
- An empty profile

### Concept check
How does `**kwargs` differ from a fixed parameter list, and why must callers provide names for these dynamic arguments?

````python
def create_profile(**kwargs):
    """Print every supplied profile attribute."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")

create_profile(name="Ashwik", age=25)
create_profile(name="Ashwik", course="Python", city="Bengaluru")
create_profile()
````

## Challenge 4 — Pass a function as an argument

### Goal
Create a function that accepts another function and uses it to perform a calculation.

### Example interface

```text
calculate(add, 10, 20)
calculate(multiply, 10, 20)
```

### Approach

- Create small operation functions such as addition and multiplication.
- Create a higher-order `calculate` function with parameters for an operation and two values.
- Call the supplied operation from inside `calculate`.
- Return the result so the caller can reuse it.
- Pass the function object by name—do not call it before passing it.

### Suggested tests
Use the same two numbers with at least two different operation functions.

### Concept check
What is the difference between passing `add` and passing `add(10, 20)`?

````python
def add(a, b):
    """Return the sum of two values."""
    return a + b

def multiply(a, b):
    """Return the product of two values."""
    return a * b

def calculate(operation, a, b):
    """Apply an operation to two values."""
    return operation(a, b)

print(calculate(add, 10, 20))
print(calculate(multiply, 10, 20))
````

## Challenge 8 — Recursive factorial

### Goal
Write a recursive function that returns the factorial of a non-negative integer.

### Approach

- Define a base case that returns a result without making another recursive call.
- Define the recursive case in terms of a smaller version of the same problem.
- Ensure every recursive call moves toward the base case.
- Handle negative inputs clearly instead of recursing forever.

### Suggested tests

- `5` → `120`
- `1` → `1`
- `0` → `1`
- A negative value → clear validation result

### Manual trace
Before running the code, expand the calls for factorial of 4 until the base case, then evaluate the returns in reverse order.

### Concept check
What would happen if the base case were missing or unreachable?

````python
def factorial(number):
    """Return the factorial of a non-negative integer."""
    if number < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    if number == 0:
        return 1
    return number * factorial(number - 1)

print(factorial(5))
print(factorial(0))
````

## Challenge 9 — Recursive sum from 1 through n

### Goal
Write a recursive function that calculates `1 + 2 + 3 + ... + n`.

### Approach

- Decide the valid input domain and handle negative values clearly.
- Define the smallest input that can be answered directly as the base case.
- Express the result for `n` using `n` plus the result of a smaller call.
- Verify that each recursive call reduces the problem toward the base case.

### Suggested tests

- `5` → `15`
- `1` → `1`
- `0` → `0` if zero is included in your accepted domain

### Manual trace
Expand the calls for `n = 4`, then show how the returned values combine.

### Concept check
Which value changes in each recursive call, and how does that guarantee termination?

````python
def recursive_sum(number):
    """Return the sum from 1 through number."""
    if number < 0:
        raise ValueError("Number must be non-negative")
    if number == 0:
        return 0
    return number + recursive_sum(number - 1)

print(recursive_sum(5))
print(recursive_sum(0))
````

## Challenge 10 — Recursive Fibonacci number

### Goal
Create a recursive function that calculates the nth Fibonacci number.

### Sequence convention
Use and state one convention consistently. A common choice is:

```text
F(0) = 0
F(1) = 1
F(n) = F(n - 1) + F(n - 2)
```

### Approach

- Validate that `n` is a non-negative integer.
- Define both required base cases.
- Build the recursive case from the two preceding positions.
- Test only modest values because the basic recursive version repeats work and becomes slow quickly.

### Suggested tests

- `F(0)` → `0`
- `F(1)` → `1`
- `F(6)` → `8`

### Manual trace
Draw the call expansion for `F(4)` and identify repeated subproblems.

### Concept check
Why does this basic implementation make far more calls than recursive factorial?

````python
def fibonacci(number):
    """Return the Fibonacci number at the given position."""
    if number < 0:
        raise ValueError("Number must be non-negative")
    if number < 2:
        return number
    return fibonacci(number - 1) + fibonacci(number - 2)

print(fibonacci(0))
print(fibonacci(1))
print(fibonacci(6))
````

## Challenge 12 — Modular mini calculator

### Goal
Create a mini calculator in which each mathematical operation is implemented as a separate function and a `main()` function controls the program.

### Recommended structure

- Create separate functions for addition, subtraction, multiplication, and division.
- Give every operation function a docstring and return its result.
- Handle a zero divisor inside or immediately around the division operation.
- Let `main()` display the menu, receive user input, route to the correct operation, display results, and control repetition or exit.
- Call `main()` once at the bottom so the program has a clear entry point.
- Keep calculations out of the menu branches; delegate them to the operation functions.

### Suggested test flow

- Complete one valid calculation for each operation.
- Attempt division by zero.
- Enter an invalid menu choice.
- Exit cleanly.

### Concept check
Which responsibilities belong to the operation functions, and which belong to `main()`? How does this separation make the program easier to extend?

````python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

def subtract(a, b):
    """Return the difference of two numbers."""
    return a - b

def multiply(a, b):
    """Return the product of two numbers."""
    return a * b

def divide(a, b):
    """Return the quotient or an error for zero division."""
    if b == 0:
        return "Cannot divide by zero"
    return a / b

def main():
    """Run the calculator menu."""
    operations = {"1": add, "2": subtract, "3": multiply, "4": divide}
    while True:
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Exit")
        choice = input("Choose an option: ")
        if choice == "5":
            break
        if choice not in operations:
            print("Invalid choice")
            continue
        first = float(input("Enter first number: "))
        second = float(input("Enter second number: "))
        print(operations[choice](first, second))

main()
````
