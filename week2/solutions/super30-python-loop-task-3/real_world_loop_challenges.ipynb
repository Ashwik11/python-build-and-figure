## Question 8 — Character frequency without `Counter`

### Problem
Count the frequency of every character in a string without using `Counter`.

### Example

**Input:** `banana`

```text
b -> 1
a -> 3
n -> 2
```

### How to approach it

- Read the string from the user.
- Create an empty dictionary for character counts.
- Iterate through the string one character at a time.
- If the character is already a dictionary key, increase its count; otherwise create it with a count of one.
- After counting, use a second loop to print every character and frequency.

### Additional test
Use a string containing a repeated space or uppercase letter and explain whether your program treats it as a separate character.


````python
input_str = "banana"
counts = {}

for character in input_str:
    if character not in counts:
        counts[character] = 0
    counts[character] += 1

for character, count in counts.items():
    print(f"{character} -> {count}")
````

## Question 9 — Second-largest number without `sort()`

### Problem
Find the second-largest number in a list without using `sort()`.

### Assumption
Treat “second-largest” as the second-largest **distinct** value. Duplicate copies of the maximum do not count as the second-largest value.

### How to approach it

- Track the largest and second-largest distinct values while iterating.
- When a new largest value is found, move the old largest into the second-largest position.
- Otherwise, update the second-largest value only when the current value lies below the largest but above the current second-largest.
- Ignore duplicate copies of the current largest.
- Handle a list that does not contain two distinct values.

### Suggested tests

- `[12, 35, 7, 24, 18]` → `24`
- `[12, 35, 7, 35, 24]` → `24`



````python
numbers = [12, 35, 7, 35, 24]
largest = None
second_largest = None

for number in numbers:
    if largest is None or number > largest:
        second_largest = largest
        largest = number
    elif number != largest and (second_largest is None or number > second_largest):
        second_largest = number

print(largest, second_largest)
````

## Question 10 — Palindrome check using loops

### Problem
Check whether a string is a palindrome using loop-based comparison.

### Comparison rule
For the basic solution, compare the entered characters exactly. If you choose to ignore letter case or spaces, state and apply that normalization clearly before the loop.

### How to approach it

- Compare characters at matching positions from the beginning and end.
- Move inward through only the first half of the string.
- If any pair differs, mark the string as not a palindrome and stop checking.
- If no mismatch is found, report that the string is a palindrome.
- Avoid solving the problem only through reverse slicing; the comparison itself should use a loop.

### Required tests

- `madam` → palindrome
- `hello` → not a palindrome
- A one-character string → palindrome



````python
input_str = input()
is_palindrome = True

for index in range(len(input_str) // 2):
    if input_str[index] != input_str[-index - 1]:
        is_palindrome = False
        break

print(is_palindrome)
````

## Question 12 — Basic ATM simulation

### Problem
Create an ATM program that repeatedly allows the user to:

1. Check balance
2. Deposit money
3. Withdraw money
4. Exit

### Recommended organization

- Store the current balance in one clearly named variable.
- Put the repeating menu inside a loop that runs until the user selects Exit.
- Use a clean conditional branch for each menu choice.
- To avoid one unreadable block, separate display, deposit, and withdrawal responsibilities into small functions if functions are permitted in your current module.
- Accept only positive deposit and withdrawal amounts.
- Reject a withdrawal that exceeds the available balance.
- Print a helpful message for an invalid menu option.
- Update the balance only after a valid transaction.

### Required test flow

**Successful cases:**

- Check the starting balance.
- Deposit a positive amount and confirm that the balance increases.
- Withdraw an affordable positive amount and confirm that the balance decreases.
- Exit and confirm that the menu stops.

**Unsuccessful cases:**

- Attempt to withdraw more than the available balance.
- Enter a zero or negative transaction amount.
- Enter an invalid menu choice.


````python
balance = 10000

while True:
    print("\n1. Check balance")
    print("2. Deposit money")
    print("3. Withdraw money")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        print(f"Current balance: {balance}")

    elif choice == "2":
        amount = float(input("Enter deposit amount: "))

        if amount > 0:
            balance += amount
            print(f"Deposit successful. New balance: {balance}")
        else:
            print("Deposit amount must be positive.")

    elif choice == "3":
        amount = float(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrawal successful. New balance: {balance}")

    elif choice == "4":
        print("Thank you for using the ATM.")
        break

    else:
        print("Invalid option. Please choose 1, 2, 3, or 4.")
````
