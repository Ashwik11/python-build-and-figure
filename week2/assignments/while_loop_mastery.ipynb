## The three parts of a safe `while` loop

Every major loop in this assignment should include short comments identifying:

1. **Initialization** — the value or state established before the loop begins.
2. **Condition** — the Boolean expression that decides whether another iteration runs.
3. **Update or termination** — the change that eventually makes the condition false, or the explicit action that exits the loop.

### How to approach it

- Repeatedly request and convert a number.
- Treat zero as a sentinel: it controls termination and must not change the total.

- First input `0` → total `0`

### Why `while` is preferable here
The number of entries is unknown. The loop must continue until a value, rather than a predetermined count, signals completion.
