# Final Combined Challenge: Loops, `while`, and Functions

Build 12 practical Python mini-programs that combine reusable functions with count-controlled and condition-controlled iteration.

Each project below includes design guidance, edge cases, and an empty implementation cell. The programs are intentionally independent so they can later be placed in separate Python files.

## Design principles

- Give each function one clear responsibility.
- Put problem-solving logic inside functions rather than calculating globally.
- Add concise docstrings to custom functions.
- Use `for` when iterating through a known collection or range.
- Use `while` when repetition depends on a changing condition or an Exit choice.
- Validate user input before changing stored data.
- Use meaningful names and keep menu branches small.
- Return reusable values from calculation functions; display them at the application boundary.
- Test successful, unsuccessful, boundary, and empty-data cases.

> The strongest implementation is not the one with the fewest lines—it is the one whose behavior is easy to understand, test, and extend.

## Suggested workflow

For each mini-project:

1. Decide how the data will be represented.
2. List the required functions and what each one returns.
3. Identify where a `for` loop or `while` loop is appropriate.
4. Implement the smallest useful function first.
5. Connect the functions through a controller or `main()` function.
6. Run the suggested test flow and inspect every state change.
7. Keep the final cell free of unhandled errors.

## Project 1 — Student Result Management System

### Goal
Create a program that accepts student marks, calculates the result, and displays a clear summary.

### Suggested function responsibilities

- Accept and validate marks.
- Calculate the total.
- Calculate the percentage.
- Assign a grade.
- Determine pass or fail.
- Format and display the final result.

### Design guidance

- Store the marks in a list.
- Use a loop to collect or process subject marks.
- Validate that every mark falls within the accepted range.
- Decide and document the grading thresholds and pass rule before coding because they are not specified.
- Return values from calculation functions so the display function can combine them.
- Keep input, calculation, and presentation responsibilities separate.

### Suggested test flow

- A student who passes every subject
- A student with a high overall grade
- A student who fails at least one subject
- A mark at every important grade boundary
- An invalid mark below 0 or above 100

### Design check
Does changing the grading scale require editing only the grade function, or does the rule appear in several places?

## Project 2 — Banking Application

### Goal
Build a menu-driven banking program with balance checking, deposits, withdrawals, transaction history, and Exit.

### Suggested function responsibilities

- Display the current balance.
- Validate and process a deposit.
- Validate and process a withdrawal.
- Display transaction history.
- Control the application menu.

### Design guidance

- Store the balance as numeric state.
- Store transaction records in a list; each record can include a type, amount, and resulting balance.
- Use a `while` loop because the user decides when the session ends.
- Accept only positive transaction amounts.
- Reject withdrawals that exceed the balance.
- Record only completed transactions.
- Return the updated balance rather than relying unnecessarily on a global variable.
- Provide a helpful response for an invalid menu choice.

### Suggested test flow

- Check the opening balance.
- Complete a deposit and withdrawal.
- Attempt an overdraw.
- Attempt a zero or negative amount.
- Inspect the transaction history.
- Enter an invalid choice, then exit cleanly.

### Design check
Can every failed operation leave both the balance and transaction history unchanged?

## Project 3 — Inventory Management

### Goal
Maintain products with a name, price, and quantity, and support common inventory operations.

### Required capabilities

- Add a product.
- Display all products.
- Search for a product.
- Update a product's quantity.
- Calculate the total inventory value.

### Design guidance

- A dictionary keyed by a normalized product name makes searching and updating direct.
- Each product record can store price and quantity.
- Decide whether adding an existing product should be rejected or treated as an update.
- Validate that price is not negative and quantity is a valid non-negative integer.
- Calculate each product's inventory value as price multiplied by quantity, then accumulate the total.
- Handle searches and updates for missing products without crashing.
- Keep menu control separate from inventory operations if you add an interactive menu.

### Suggested test flow

- Add two valid products.
- Attempt a duplicate product.
- Search for an existing and missing product.
- Update quantity to a valid value and to zero.
- Reject negative price or quantity.
- Verify the total inventory value manually.

### Design check
Does product-name normalization prevent accidental duplicates such as `Mouse` and `mouse`?

## Project 4 — Quiz Application

### Goal
Create a quiz with at least five Python questions, one-at-a-time display, answer checking, score tracking, and a final percentage.

### Suggested data structure
Use a list of question records. Each record can contain the question text and expected answer, keeping content separate from quiz logic.

### Suggested function responsibilities

- Present one question and collect an answer.
- Normalize and check the answer.
- Run the complete quiz.
- Calculate the final percentage.
- Display the result.

### Design guidance

- Use a loop to process the question collection.
- Increase the score only for a correct response.
- Normalize surrounding spaces and letter case when appropriate.
- Calculate percentage from score and question count.
- Avoid hard-coding the denominator so more questions can be added later.
- Consider immediate feedback after each answer.

### Suggested test flow

- All correct answers
- A mixture of correct and incorrect answers
- Answers with different capitalization or surrounding spaces
- Verify that five correct answers out of five produces 100%

### Design check
Can a new question be added by changing only the question data rather than the quiz-control logic?

## Project 5 — Number Analysis Tool

### Goal
Create a function that accepts a list and returns:

- Largest number
- Smallest number
- Total
- Average
- Even count
- Odd count
- Positive count
- Negative count

Do not use `min()`, `max()`, or `sum()`.

### Design guidance

- Decide how an empty list will be handled before reading its first element.
- Initialize largest and smallest from an actual list value.
- Initialize the total and four counters.
- Analyze all values in one loop.
- Treat zero as even, but neither positive nor negative.
- Calculate the average only after the total and count are known.
- Return all results in a structured form such as a dictionary.

### Suggested tests

- A mixture of positive, negative, even, odd, and zero values
- A list containing one value
- A list containing only negative numbers
- An empty list

### Design check
Can the list be traversed once while still producing every required statistic?

## Project 8 — Password Strength Checker

### Goal
Create a function that checks whether a password contains:

- At least one uppercase letter
- At least one lowercase letter
- At least one number
- At least one special character
- At least eight characters

Return a meaningful result.

### Design guidance

- Initialize one Boolean flag for each character requirement.
- Iterate through the password once and update the flags.
- Treat a special character as a character that is not a letter or digit, or use a clearly defined punctuation set.
- Check the length requirement separately.
- Return a structured result, a list of missing requirements, or a descriptive strength message.
- Do not print the password itself in diagnostic output.

### Suggested tests

- A password satisfying all five requirements
- A password shorter than eight characters
- Missing uppercase, lowercase, number, and special-character cases
- An empty string

### Design check
Can the caller tell exactly which requirements failed instead of receiving only `Weak`?

## Project 9 — Prime Number Analyzer

### Goal
Accept two numbers defining a range, then find all primes, count them, calculate their sum, and display the largest prime found.

### Suggested function responsibilities

- Determine whether one number is prime.
- Find all primes in an inclusive range.
- Count the prime list.
- Calculate the prime total.
- Find the largest discovered prime.
- Display the analysis.

### Design guidance

- Treat values below 2 as non-prime.
- Normalize reversed boundaries or reject them with a clear message.
- Reuse the single-number prime checker instead of duplicating divisor logic.
- Avoid `sum()` and `max()` if you want to reinforce manual loop analysis.
- Handle a range containing no primes before requesting the largest prime.

### Suggested tests

- Range 1 through 10 → primes `2, 3, 5, 7`, count `4`, total `17`, largest `7`
- A single-value prime range
- A range with no primes
- Reversed boundaries
- Negative boundaries

### Design check
Is primality logic written once and reused by the range function?

## Project 11 — Mini Authentication System

### Goal
Create a small application with predefined credentials, limited login attempts, success and failure handling, logout, and retry logic.

### Suggested function responsibilities

- Validate one username-password pair.
- Run a login sequence with a maximum attempt count.
- Run the authenticated session.
- Log out and return to the appropriate state.
- Control whether another login sequence begins.

### Design guidance

- Store only the predefined test credentials needed for this learning project.
- Track attempts explicitly and never exceed the configured maximum.
- Count a failed credential pair as one attempt.
- Start the authenticated menu only after a successful login.
- Ensure logout changes the authentication state before retrying.
- Decide whether exhausting attempts ends the application or offers a clearly controlled restart.
- Avoid displaying the expected password in failure messages.

### Suggested test flow

- Correct credentials on the first attempt
- Incorrect password followed by success
- Incorrect username and password until attempts are exhausted
- Successful login followed by logout and another login

### Design check
Are the logged-out, authenticating, authenticated, and locked-or-ended states clearly separated?

## Project 12 — Python Utility Application

### Goal
Create one menu-driven application containing at least five reusable utilities.

### Possible utilities

- Calculator
- Palindrome checker
- Prime checker
- Factorial calculator
- Multiplication table generator
- Number analyzer
- Password checker
- Unit converter
- Text statistics

### Recommended structure

- Implement every utility as its own function or small group of focused functions.
- Create one menu that maps choices to utilities.
- Use a `while` loop so the application continues until Exit.
- Keep the menu controller responsible for navigation, not individual calculations.
- Validate input within the appropriate utility.
- Return to the main menu after each completed or rejected operation.
- Provide an invalid-choice response without terminating the application.
- Add your own feature only if you can describe and test every line.

### Suggested test flow

- Run all five selected utilities successfully.
- Test one important failure or boundary case for each utility.
- Enter an invalid main-menu choice.
- Confirm that each utility returns to the menu.
- Exit cleanly.

### Design check
Can a sixth utility be added with a new function and one menu route without rewriting the existing utilities?

## Final quality checklist

### Structure and readability

- [ ] All 12 mini-programs can run independently.
- [ ] Custom functions have concise docstrings.
- [ ] Names describe purpose rather than implementation details.
- [ ] Menu controllers delegate calculations to helper functions.
- [ ] `for` and `while` are selected according to the type of repetition.
- [ ] Comments clarify non-obvious decisions without narrating every line.

### Correctness and safety

- [ ] Numeric inputs are validated before calculations or state changes.
- [ ] Empty collections are handled before indexing, division, or largest-value checks.
- [ ] Zero and boundary values have deliberate behavior.
- [ ] Failed banking, cart, inventory, and expense operations preserve existing data.
- [ ] Interactive programs always provide a reachable Exit path.
- [ ] Division by zero and insufficient funds are handled.
- [ ] Authentication attempts cannot exceed the configured limit.
- [ ] Password values are not exposed in output.

### Testing

- [ ] Every program has at least one normal test.
- [ ] Every program has at least one boundary or unsuccessful test.
- [ ] Expected state changes have been calculated manually.
- [ ] All cells run from top to bottom without unhandled errors.
- [ ] Every line is understood and can be described in your own words.
