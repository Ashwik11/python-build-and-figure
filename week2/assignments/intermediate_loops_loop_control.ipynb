## Question 11 — Determine whether a number is prime

### Problem
Use a loop to determine whether a user-provided integer is prime.

### Prime-number rule
A prime number is greater than 1 and has exactly two positive factors: 1 and itself.

### How to approach it

- Handle values less than 2 as not prime.
- Test possible divisors using a loop.
- If any divisor produces no remainder, the number is not prime and the search can stop.
- If the loop completes without finding a divisor, the number is prime.
- A `for-else` structure is a natural fit, though a clearly explained flag-based solution can also work.

### Sample tests

- `2` → prime
- `17` → prime
- `21` → not prime
- `1` → not prime
