# Python Functions Fundamentals



## Core vocabulary for your video

- **Function definition:** The `def` statement that creates a reusable block of behavior.
- **Function call:** The expression that runs the function.
- **Parameter:** A variable named in the function definition.
- **Argument:** A value supplied when calling the function.
- **Return value:** A result sent back to the caller with `return`.
- **Default argument:** A parameter value used when the caller does not provide one.
- **Positional argument:** An argument matched according to its position.
- **Keyword argument:** An argument matched by explicitly naming its parameter.
- **Local scope:** Variables created inside a function normally exist only during that function call.

> **Required rule:** Every custom function must contain a concise docstring immediately below its definition.

## Question 4 — Largest of three numbers without `max()`

### Problem
Create a function that returns the largest of three numbers without using `max()`.

### How to approach it

- Accept three numeric parameters.
- Compare the values inside the function.
- Either track a current largest value or use a complete conditional chain.
- Make sure equal values are handled correctly.
- Return the largest number.

### Suggested tests

- `(10, 25, 7)` → `25`
- `(30, 30, 12)` → `30`
- `(-8, -2, -5)` → `-2`



````python
def largest_of_three(a, b, c):
    largest = a

    if b > largest:
        largest = b
    if c > largest:
        largest = c

    return largest

# Suggested tests
print(largest_of_three(10, 25, 7))
print(largest_of_three(30, 30, 12))
print(largest_of_three(-8, -2, -5))
````

## Question 5 — Factorial function

### Problem
Create a function that calculates and returns the factorial of a number.

### How to approach it

- Accept one non-negative integer parameter.
- Keep the accumulator and loop inside the function.
- Initialize the product using the multiplicative identity.
- Multiply through the required factors.
- Return the completed product.
- Handle negative input clearly because factorial is not defined for negative integers in this assignment.
- Remember that `0!` equals `1`.

### Suggested tests

- `5` → `120`
- `1` → `1`
- `0` → `1`
- A negative value → clear validation result



````python
def factorial(n):
    if n < 0:
        return "Error: Factorial is not defined for negative numbers"
    if n == 0 or n == 1:
        return 1
    
    product = 1
    for i in range(2, n + 1):
        product *= i
    return product

# Suggested tests
print(factorial(5))
print(factorial(1))
print(factorial(0))
print(factorial(-3))
print(factorial(number))
````

## Question 6 — Prime-number function

### Problem
Create a function that checks whether a number is prime.

### Prime rule
A prime number is greater than 1 and has exactly two positive factors: 1 and itself.

### How to approach it

- Accept one integer parameter.
- Return a Boolean value so the caller can reuse the decision.
- Treat all values below 2 as not prime.
- Search for a divisor inside the function.
- Return `False` as soon as a valid divisor is found.
- Return `True` only after all necessary checks succeed.

### Suggested tests

- `2` → `True`
- `17` → `True`
- `21` → `False`
- `1` → `False`
- `-5` → `False`



````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Test the function
print(is_prime(2))
print(is_prime(17))
print(is_prime(21))
print(is_prime(1))
print(is_prime(-5))
print(is_prime(number))
````

## Question 8 — Sum a list without `sum()`

### Problem
Create a function that accepts a list and returns its total without using `sum()`.

### How to approach it

- Accept the list as a parameter.
- Initialize a local total inside the function.
- Iterate through the list and add each value to the local total.
- Return the total after the loop.
- Do not rely on a total variable defined outside the function.

### Suggested tests

- `[10, 20, 30]` → `60`
- `[-5, 2, 3]` → `0`
- `[]` → `0`



````python
def sum_list(values):
    total = 0
    for value in values:
        total += value
    return total

# Suggested tests
print(sum_list([10, 20, 30]))
print(sum_list([-5, 2, 3]))
print(sum_list([]))
````

## Question 9 — Count vowels

### Problem
Create a function that accepts a string and returns the number of vowels.

### How to approach it

- Accept the text as a parameter.
- Initialize a local counter.
- Iterate through the characters inside the function.
- Compare case-insensitively with the vowels `a`, `e`, `i`, `o`, and `u`.
- Increase the counter for each match and return it after the loop.

### Suggested tests

- `"Education"` → `5`
- `"PYTHON"` → `1`
- `"rhythm"` → `0`
- An empty string → `0`



````python
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for char in text.lower():
        if char in vowels:
            count += 1

    return count

# Suggested tests
print(count_vowels("Education"))
print(count_vowels("PYTHON"))
print(count_vowels("rhythm"))
print(count_vowels(""))
````

## Question 10 — Palindrome function

### Problem
Create a function that accepts a string and determines whether it is a palindrome.

### Comparison rule
For the basic version, compare characters exactly. If you choose to ignore letter case or spaces, normalize the text inside the function and clearly explain that rule.

### How to approach it

- Accept the string as a parameter.
- Compare matching characters from the beginning and end using loop-based logic.
- Return `False` when a mismatch is found.
- Return `True` after every required pair matches.
- Keep both normalization and comparison inside the function.

### Suggested tests

- `"madam"` → `True`
- `"hello"` → `False`
- A one-character string → `True`
- An empty string → `True` under the usual definition



````python
def is_palindrome(text):
    normalized = text.lower().replace(" ", "")

    left = 0
    right = len(normalized) - 1

    while left < right:
        if normalized[left] != normalized[right]:
            return False
        left += 1
        right -= 1

    return True

# Suggested tests
print(is_palindrome("madam"))
print(is_palindrome("hello"))
print(is_palindrome("a"))
print(is_palindrome(""))
print(is_palindrome(str(number)))
````

## Question 12 — `print()` versus `return`

### Problem
Demonstrate the difference between a function that uses `print()` and one that uses `return`. Explain through code why `return` is useful.

### How to approach it

- Create two simple functions that perform the same calculation.
- In the first function, print the result but do not explicitly return it.
- In the second function, return the result.
- Capture the outcome of each call in a variable.
- Display the captured values to show that a print-only function implicitly returns `None`.
- Reuse the returned numeric value in another calculation to demonstrate composability.
- Give both functions docstrings.



````python
def print_result(a, b):
    """Prints the sum but returns None implicitly."""
    print(a + b)

def return_result(a, b):
    """Returns the sum for reuse in other calculations."""
    return a + b

# Capture outcomes
outcome_print = print_result(5, 3)
outcome_return = return_result(5, 3)

print(f"print_result() returned: {outcome_print}")
print(f"return_result() returned: {outcome_return}")

# Demonstrate composability - only return works
doubled = return_result(5, 3) * 2
print(f"Doubled result: {doubled}")

# Attempting this with print_result would fail
try:
    doubled_print = print_result(5, 3) * 2
except TypeError as e:
    print(f"Error with print-only function: {e}")
````
